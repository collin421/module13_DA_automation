# **문자열 함수**

#### **1. len() 함수**
- len() 함수는 문자열의 길이를 반환함.

```python
강의 = "금융데이터 분석 강의"
상품명_길이 = len(강의)
print(상품명_길이)
```

```python
고객이름 = "홍길동"
이름길이 = len(고객_이름)
이름길이
```

#### **2. index() 함수**
- 문자열에서 특정 문자가 처음 등장하는 위치를 반환.
- 찾는 값이 없을 경우 error

```python
상품_코드 = "FUND_2024_주식형"
```

```python
위치 = 상품_코드.index('주')
위치
```

```python
상품_코드 = "FUND_2024_주식형"
위치 = 상품_코드.index('2024')
위치
```

#### **3. find() 함수**
- 문자열에서 특정 문자가 처음 등장하는 위치.
- 찾는 값이 없을 경우 -1 

```python
보고서_제목 = "2024년 투자 전략 보고서"
```

```python
위치 = 보고서_제목.find('전략')
위치
```

```python
위치 = 보고서_제목.find('5')
위치
```

#### **4. count() 함수**
- 문자열에서 특정 문자가 몇 번 등장하는지를 반환.

```python
내용 = "주식, 주식, 채권, 주식, 채권, ETA"
```

```python
주식_개수 = 리포트_내용.count('주식')
주식_개수
```

#### **5. upper(), lower() 함수**
- upper() 함수는 문자열을 대문자로 변환함.
- lower() 함수는 문자열을 소문자로 변환함.

```python
상품코드 = "Bond_ETF_2024"
```

```python
대문자 = 상품코드.upper()
대문자
```

```python
소문자 = 상품코드.lower()
소문자
```

#### **6. strip() 함수**
-  문자열의 양쪽 공백을 제거

```python
종목코드 = "     125534     "
print(종목코드.strip())
```

```python
거래내역 = "   매수 주문   "
print(거래내역.strip())
```

#### **7. split() 함수**
-  문자열을 특정 구분자를 기준으로 분할

```python
상품군 = "주식/ETF/채권"
상품분활 = 상품군.split("/")
print(상품분활)
```

```python
날짜 = "2024-03-24"
날짜.split("-")
```

```python
날짜 = "    2024-03-24     "
날짜.strip().split("-")
```

#### **8. join() 함수**
-  문자열을 특정 구분자로 결합하여 하나의 문자열로 만듦

```python
상품1 = "주식"
상품2 = "ETF"
상품3 = "2024"
결합상품 = "-".join([상품1, 상품2, 상품3])

결합상품
```

- 문자열은 immutable 데이터
- 변경할 수 없은 데이터
- replace 함수 또는 슬라이싱, 인덱싱을 통해 새로운 변수 할당 

```python
종목 = "kakao"
종목[0] = "T"
```

#### **9. replace() 함수**
-  문자열의 특정 부분을 다른 문자열로 대체

```python
상품명 = "채권투자_2023"
새상품명 = 상품명.replace("2023", "2024")
새상품명
```

```python
종목명 = "kakao"
새종목명 = "T" + 종목명[1:]
새종목명
```

```python
종목명 = "kakao"
새종목명 = 종목명.replace('k', 'T')
새종목명2 = 종목명.replace('k', 'T', 1)
print(새종목명)
print(새종목명2)
```
