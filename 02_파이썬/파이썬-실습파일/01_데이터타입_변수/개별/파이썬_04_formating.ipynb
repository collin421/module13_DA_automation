# 포맷팅
- 데이터를 특정 형식으로 출력
- format() 함수는 문자열 안에 변수나 값을 삽입할 때 사용

### **1. 기본 포맷팅**
- {} 안에 값이 들어가며, 여러 값을 순서대로 삽입할 수 있음

````python
상품 = "주식"
가격 = 10000
print("상품명: {}, 가격: {}원".format(상품, 가격))
````

- 인덱스 번호를 중괄호 안에 넣어 포맷팅 순서를 제어

````python
상품 = "채권"
가격 = 15000
print("상품명: {0}, 가격: {1}원".format(상품, 가격))
print("가격: {1}원, 상품명: {0}".format(상품, 가격))
````

- 포맷팅 시 변수 이름을 지정하여 가독성을 높일 수 있음

````python
print("상품명: {상품}, 가격: {가격}원".format(상품="ETF", 가격=20000))
````

### **2. F-string**
- f-string은 가장 최신의 포맷팅 방식으로, 문자열 앞에 **f**를 붙이고, 중괄호 {} 안에 변수를 직접 삽입함.
- 가독성이 좋고 간결하게 표현할 수 있음.
- Python 3.6 이상 사용가능

````python
상품 = "주식"
가격 = 15000
print(f"상품명: {상품}, 가격: {가격}원")
````

### **2. 숫자 포맷팅**

#### 소수점 자리수 지정 :
- f"{변수명:포맷}"
- {변수:.nf} 형식으로 소수점 자리수를 지정함.
- 예시 {:.2f}는 소수점 아래 두 자리까지 출력함.

````python
이자율 = 0.034567
print("이자율: {:.2f}%".format(이자율 * 100))
````

````python
# f-string
print(f"이자율 : {이자율*100:.2f}%")
````

````python
나누기 = 10/3 
print("10나누기 3은 {:.2f}입니다".format(나누기))
````

````python
# f-string
print(f"10나누기 3은 {나누기:.2f}입니다")
````

#### 천단위 구분 기호
- 큰 숫자에 **콤마(,)**를 추가해 가독성을 높일 수 있음.
- {변수:,} 형식을 사용해 1,000 단위로 콤마를 삽입함.

````python
금액 = 10000000
print("금액: {:,}원".format(금액))
````

````python
# f-string
금액 = 10000000
print(f"금액: {금액:,}원")
````

### 소수점 자릿수와 콤마를 함께 사용

````python
수익 = 1234567.89123
print("수익금: {:,.2f}원".format(수익))
print(f"수익금: {수익:,.2f}원")
````

### **3. 날짜 포맷팅**

datetime 모듈 : 는 날짜와 시간을 표현하는 데 사용
- date: 연도, 월, 일을 저장하고 관리
- time: 시간, 분, 초, 마이크로초를 저장하고 관리
- datetime: 날짜와 시간 정보를 모두 포함
- datetime.strptime() :문자열을 datetime 형식으로 바뀜

**날짜 시간 표시 형식**
- %Y: 4자리 연도 (예: 2024)
- %m: 2자리 월 (01~12)
- %d: 2자리 일 (01~31)
- %H: 24시간 형식의 시간 (00~23)
- %M: 분 (00~59)
- %S: 초 (00~59
- %A: 요일의 풀네임 (예: Monday, Tuesday)
- %a: 요일의 약자 (예: Mon, Tue)

````python
# datetime 모듈 임포트
from datetime import datetime

# 현재 날짜와 시간 얻기
날짜시간 = datetime.now()
````

````python
# %Y 연도, %m 월, %d 일
print(f"{날짜시간: %Y연 %m월 %d일}")

# %H 시간, %M 분, %S 초
print(f"{날짜시간: %H시 %M분 %S초}")

# %A 요일 풀 네임, %a 요일의 약자
print(f"{날짜시간: %A, %a}")
````

````python
from datetime import date, time, datetime

# 특정 날짜 생성
d = date(2024, 10, 31)
print(d)

# 특정 시간 생성
t = time(14, 30, 25)
print(t)

# 특정 날짜와 시간 생성
dt = datetime(2024, 10, 31, 14, 30, 25)
print(dt)
````

````python
# 문자열을 datetime 객체로 변환
date_string = "2024-10-31"
date_object = datetime.strptime(date_string, "%Y-%m-%d")
print(date_object)
````
