## **1. 문자열 기본 표현**

**개념**
- 파이썬에서 문자열은 작은따옴표(' ') 또는 큰따옴표(" ")로 감쌈.
- 큰따옴표나 작은따옴표 중 하나를 선택해서 문자열을 정의할 수 있음.

```python
# 문자열 선언
문자열1 = "안녕하세요"
문자열2 = 'Python을 배워요'

print(문자열1)  # 출력: 안녕하세요
print(문자열2)  # 출력: Python을 배워요
```

#### **큰따옴표 안에 작은따옴표 사용**

```python
# 작은따옴표가 포함된 문자열
문자열 = "그는 '파이썬'을 좋아해요"
print(문자열)  # 출력: 그는 '파이썬'을 좋아해요
```

#### **여러 줄 문자열 (Multi-line String)**
- 세 개의 따옴표(''' ''' 또는 """ """)를 사용하여 여러 줄로 이루어진 문자열을 선언

```python
# 여러 줄 문자열
문자열 = """안녕하세요
파이썬 강의를
수강하고 있습니다."""

print(문자열)
```

#### **반복 및 연결**
- 문자열 반복: 문자열을 여러 번 반복할 때는 곱셈(*) 연산자를 사용함
- 문자열 연결: 두 개 이상의 문자열을 연결할 때는 더하기(+) 연산자를 사용함
- **문자열에서 -, / 연산자는 에러 발생함**

```python
# 문자열 반복
반복된문자열 = "Python! " * 3
print(반복된문자열)
```

```python
# 문자열 연결
문자열1 = "안녕하세요"
문자열2 = "파이썬"
연결된문자열 = 문자열1 + ", " + 문자열2 + "을 배우세요!"

print(연결된문자열)
```

- 숫자가 포함된 문자열을 문자형 숫자라고 함. 숫자처럼 보이지만 실제로는 문자로 취급됨.
- 문자형 숫자는 숫자로 변환하지 않으면 수학적 연산을 할 수 없음.

```python
# 문자형 숫자
숫자문자열 = "12345"

# 인덱싱으로 문자 하나 가져오기
print(숫자문자열[0])  # 출력: 1
print(숫자문자열[1])  # 출력: 2

# 문자형 숫자를 숫자로 변환 후 연산
숫자 = int(숫자문자열)  # 문자열을 정수로 변환
합계 = 숫자 + 100
print(합계)
```

## **3. 문자열 인덱싱과 슬라이싱**

**개념**

- 인덱싱은 문자열에서 특정 위치의 문자를 가져오는 방법임.
- 문자열은 각각의 문자에 번호(인덱스)가 부여되어 있음.
- 파이썬의 문자열 인덱스는 0부터 시작하며, 음수 인덱스를 사용해 뒤에서부터 셀 수도 있음.

**인덱스 규칙**

- 양의 인덱스는 왼쪽에서부터 시작, 0이 첫 번째 문자.
- 음의 인덱스는 오른쪽에서부터 시작, -1이 마지막 문자

```python
# 문자열 선언
문자열 = "Python"

# 양수 인덱스 사용
print(문자열[0])
print(문자열[1])
print(문자열[2])
print(문자열[3])
print(문자열[4])
print(문자열[5])
```

```python
# 음수 인덱스 사용
print(문자열[-1])
print(문자열[-2])
print(문자열[-3])
print(문자열[-4])
print(문자열[-5])
print(문자열[-6])
```

### **슬라이싱**

**개념** 
- 슬라이싱은 문자열에서 일정 범위의 부분 문자열을 추출하는 방법임.
- [시작:끝] 형태로 사용되며, 시작 인덱스부터 끝 인덱스 직전까지의 문자를 가져옴.
- 슬라이싱은 인덱스를 사용해 문자열의 일부를 자르는 작업임.

**슬라이싱 규칙**
- 문자열[START:END]: 시작 인덱스에서 끝 인덱스 직전까지 문자를 가져옴.
- 문자열[:END]: 처음부터 끝 인덱스 직전까지 가져옴.
- 문자열[START:]: 시작 인덱스부터 끝까지 가져옴.
- 문자열[START:END:STEP]: 시작 인덱스부터 끝 인덱스 직전까지 간격을 주어 가져옴.

```python
# 문자열 선언
문자열 = "Hello, Python!"

# 슬라이싱 사용
print(문자열[0:5])  # 0번 인덱스부터 4번 인덱스까지
```

```python
print(문자열[7:13])  # 7번 인덱스부터 12번 인덱스까지
```

```python
print(문자열[:5])   # 처음부터 4번 인덱스까지
```

```python
print(문자열[7:])   # 7번 인덱스부터 끝까지
```

```python
print(문자열[::2])  # 0번 인덱스부터 2칸씩 건너뛰며 가져옴
```

- 슬라이싱에서 인덱스가 범위를 벗어나도 오류가 발생하지 않으며, 가능한 범위만 가져옴

```python
# 문자열 선언
문자열 = "Python"

# 인덱스를 범위 초과했을 때 슬라이싱
print(문자열[0:100])
```

- 슬라이싱을 활용해 문자열을 역순으로 뒤집을 수 있음

```python
# 문자열 선언
문자열 = "Python"

# 문자열 뒤집기
뒤집힌문자열 = 문자열[::-1]

# 출력
print(뒤집힌문자열)
```
