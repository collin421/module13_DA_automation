# **02. 변수란 무엇인가?**

#### **개념**
- 변수는 데이터를 저장하는 이름표임.
- 프로그래밍에서 변수를 사용하면 데이터를 저장하고, 그 데이터를 나중에 사용할 수 있음.
- **변수 선언 : 변수명 = 값 형태로 사용함. =는 "할당"을 의미하며, 오른쪽 값을 왼쪽 변수에 저장함.**

#### **원리**
- 변수를 사용하면 한 번 저장한 데이터를 재사용할 수 있음.
- 프로그램에서 데이터를 저장해두고 필요할 때 불러와 사용하는 방식임.
- **변수에 저장된 데이터는 언제든지 다른 값으로 변경될 수 있음**

#### **이유**
- 변수는 데이터를 저장하고 효율적으로 사용할 수 있도록 도와줌. 반복적으로 사용되는 데이터를 중앙에서 관리할 수 있음.
- 변수를 사용하면 코드가 간결해지고, 동일한 값을 여러 번 입력하는 수고를 덜 수 있음

#### **변수 할당**

````python
# 변수 선언 및 할당
나이 = 25  # 25라는 값을 '나이' 변수에 저장
키 = 175.3 # 175.3이라는 값을 키 변수에 저장
이름 = "홍길동"  # '홍길동'이라는 문자열을 '이름' 변수에 저장

print(나이)
print(키)
print(이름)
````

````python
# 할당한 데이터에 따라서 변수의 데이터 타입 결정됨
print(type(나이))
print(type(키))
print(type(이름))
````

## **3. 문자열 인덱싱과 슬라이싱**

**개념**

- 인덱싱은 문자열에서 특정 위치의 문자를 가져오는 방법임.
- 문자열은 각각의 문자에 번호(인덱스)가 부여되어 있음.
- 파이썬의 문자열 인덱스는 0부터 시작하며, 음수 인덱스를 사용해 뒤에서부터 셀 수도 있음.

**인덱스 규칙**

- 양의 인덱스는 왼쪽에서부터 시작, 0이 첫 번째 문자.
- 음의 인덱스는 오른쪽에서부터 시작, -1이 마지막 문자

````python
# 문자열 선언
문자열 = "Python"

# 양수 인덱스 사용
print(문자열[0])
print(문자열[1])
print(문자열[2])
print(문자열[3])
print(문자열[4])
print(문자열[5])
````

````python
# 음수 인덱스 사용
print(문자열[-1])
print(문자열[-2])
print(문자열[-3])
print(문자열[-4])
print(문자열[-5])
print(문자열[-6])
````

### **슬라이싱**

**개념** 
- 슬라이싱은 문자열에서 일정 범위의 부분 문자열을 추출하는 방법임.
- [시작:끝] 형태로 사용되며, 시작 인덱스부터 끝 인덱스 직전까지의 문자를 가져옴.
- 슬라이싱은 인덱스를 사용해 문자열의 일부를 자르는 작업임.

**슬라이싱 규칙**
- 문자열[START:END]: 시작 인덱스에서 끝 인덱스 직전까지 문자를 가져옴.
- 문자열[:END]: 처음부터 끝 인덱스 직전까지 가져옴.
- 문자열[START:]: 시작 인덱스부터 끝까지 가져옴.
- 문자열[START:END:STEP]: 시작 인덱스부터 끝 인덱스 직전까지 간격을 주어 가져옴.

````python
# 문자열 선언
문자열 = "Hello, Python!"

# 슬라이싱 사용
print(문자열[0:5])  # 0번 인덱스부터 4번 인덱스까지
````

````python
print(문자열[7:13])  # 7번 인덱스부터 12번 인덱스까지
````

````python
print(문자열[:5])   # 처음부터 4번 인덱스까지
````

````python
print(문자열[7:])   # 7번 인덱스부터 끝까지
````

````python
print(문자열[::2])  # 0번 인덱스부터 2칸씩 건너뛰며 가져옴
````

- 슬라이싱에서 인덱스가 범위를 벗어나도 오류가 발생하지 않으며, 가능한 범위만 가져옴

````python
# 문자열 선언
문자열 = "Python"

# 인덱스를 범위 초과했을 때 슬라이싱
print(문자열[0:100])
````

- 슬라이싱을 활용해 문자열을 역순으로 뒤집을 수 있음

````python
# 문자열 선언
문자열 = "Python"

# 문자열 뒤집기
뒤집힌문자열 = 문자열[::-1]

# 출력
print(뒤집힌문자열)
````

# 포맷팅
- 데이터를 특정 형식으로 출력
- format() 함수는 문자열 안에 변수나 값을 삽입할 때 사용

### **1. 기본 포맷팅**
- {} 안에 값이 들어가며, 여러 값을 순서대로 삽입할 수 있음

````python
상품 = "주식"
가격 = 10000
print("상품명: {}, 가격: {}원".format(상품, 가격))
````

- 인덱스 번호를 중괄호 안에 넣어 포맷팅 순서를 제어

````python
상품 = "채권"
가격 = 15000
print("상품명: {0}, 가격: {1}원".format(상품, 가격))
print("가격: {1}원, 상품명: {0}".format(상품, 가격))
````

- 포맷팅 시 변수 이름을 지정하여 가독성을 높일 수 있음

````python
print("상품명: {상품}, 가격: {가격}원".format(상품="ETF", 가격=20000))
````

### **2. F-string**
- f-string은 가장 최신의 포맷팅 방식으로, 문자열 앞에 **f**를 붙이고, 중괄호 {} 안에 변수를 직접 삽입함.
- 가독성이 좋고 간결하게 표현할 수 있음.
- Python 3.6 이상 사용가능

````python
상품 = "주식"
가격 = 15000
print(f"상품명: {상품}, 가격: {가격}원")
````

### **2. 숫자 포맷팅**

#### 소수점 자리수 지정 :
- f"{변수명:포맷}"
- {변수:.nf} 형식으로 소수점 자리수를 지정함.
- 예시 {:.2f}는 소수점 아래 두 자리까지 출력함.

````python
이자율 = 0.034567
print("이자율: {:.2f}%".format(이자율 * 100))
````

````python
# f-string
print(f"이자율 : {이자율*100:.2f}%")
````

````python
나누기 = 10/3 
print("10나누기 3은 {:.2f}입니다".format(나누기))
````

````python
# f-string
print(f"10나누기 3은 {나누기:.2f}입니다")
````

#### 천단위 구분 기호
- 큰 숫자에 **콤마(,)**를 추가해 가독성을 높일 수 있음.
- {변수:,} 형식을 사용해 1,000 단위로 콤마를 삽입함.

````python
금액 = 10000000
print("금액: {:,}원".format(금액))
````

````python
# f-string
금액 = 10000000
print(f"금액: {금액:,}원")
````

### 소수점 자릿수와 콤마를 함께 사용

````python
수익 = 1234567.89123
print("수익금: {:,.2f}원".format(수익))
print(f"수익금: {수익:,.2f}원")
````

### **3. 날짜 포맷팅**

datetime 모듈 : 는 날짜와 시간을 표현하는 데 사용
- date: 연도, 월, 일을 저장하고 관리
- time: 시간, 분, 초, 마이크로초를 저장하고 관리
- datetime: 날짜와 시간 정보를 모두 포함
- datetime.strptime() :문자열을 datetime 형식으로 바뀜

**날짜 시간 표시 형식**
- %Y: 4자리 연도 (예: 2024)
- %m: 2자리 월 (01~12)
- %d: 2자리 일 (01~31)
- %H: 24시간 형식의 시간 (00~23)
- %M: 분 (00~59)
- %S: 초 (00~59
- %A: 요일의 풀네임 (예: Monday, Tuesday)
- %a: 요일의 약자 (예: Mon, Tue)

````python
# datetime 모듈 임포트
from datetime import datetime

# 현재 날짜와 시간 얻기
날짜시간 = datetime.now()
````

````python
# %Y 연도, %m 월, %d 일
print(f"{날짜시간: %Y연 %m월 %d일}")

# %H 시간, %M 분, %S 초
print(f"{날짜시간: %H시 %M분 %S초}")

# %A 요일 풀 네임, %a 요일의 약자
print(f"{날짜시간: %A, %a}")
````

````python
from datetime import date, time, datetime

# 특정 날짜 생성
d = date(2024, 10, 31)
print(d)

# 특정 시간 생성
t = time(14, 30, 25)
print(t)

# 특정 날짜와 시간 생성
dt = datetime(2024, 10, 31, 14, 30, 25)
print(dt)
````

````python
# 문자열을 datetime 객체로 변환
date_string = "2024-10-31"
date_object = datetime.strptime(date_string, "%Y-%m-%d")
print(date_object)
````
