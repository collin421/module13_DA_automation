### **1) 함수란**
- 동작을 미리 정의해두고, 필요할 때 그 동작을 반복해서 사용할 수 있게 해주는 도구임.
- 마치 레시피처럼, 함수를 한 번 만들어 놓으면 같은 작업을 여러 번 반복할 때마다 다시 적지 않고, 필요할 때 호출해서 사용할 수 있음.



#### **함수의 장점**
- **1. 코드의 재사용성**
    - 함수를 사용하면 한 번 정의한 동작을 여러 번 재사용할 수 있음.
    - 같은 기능을 하는 코드를 반복해서 작성할 필요가 없어짐.
    - 예)이자율 계산을 해야 한다고 가정하면, 매번 계산 공식을 적지 않고 함수로 한 번 정의해두고 필요할 때마다 그 함수를 불러오면 됨
- **2. 코드의 가독성 향상**
    - 코드가 더 짧고 명확해짐.
    - 코드를 읽을 때 함수 이름만 봐도 그 함수가 어떤 동작을 수행하는지 쉽게 알 수 있음.
- **3. 유지보수 용이성**
    - 코드의 중복을 줄이기 때문에 나중에 코드가 변경되더라도 한 번에 쉽게 수정할 수 있음.
    - 만약 함수에서 사용된 동작을 수정해야 한다면, 함수 정의 부분만 수정하면 됨.

#### **함수 없는 코드**
- 함수를 사용하지 않으면, 같은 계산을 여러 번 직접 작성해야 함.

```python
# 5년 동안 원금 1000만원, 연 이자율 5%를 적용해 원리금을 계산

# 첫번째 고객의 원금, 이자율, 기간
원금 = 10000000
이자율 = 0.05
기간 = 5

# 첫 번째 이자 계산
원리금_1 = 원금 + (원금 * 이자율 * 기간)
print(f"1차 원리금: {원리금_1}원")
```

```python
# 두번째 고객
원금 = 20000000
이자율 = 0.03
기간 = 3

# 두 번째 이자 계산 (다시 작성해야 함)
원리금_2 = 원금 + (원금 * 이자율 * 기간)
print(f"2차 원리금: {원리금_2}원")
```

#### **함수 있는 코드**
- 같은 계산을 함수로 정의하면, 코드를 반복해서 작성할 필요가 없음. 함수만 한 번 호출하면 됨

```python
# 이자율 계산 함수 정의
def 원리금_계산(원금, 이자율, 기간):
    return 원금 + (원금 * 이자율 * 기간)

# 함수 호출로 원리금 계산
원리금_1 = 원리금_계산(10000000, 0.05, 5)
print(f"1차 원리금: {원리금_1}원")

원리금_2 = 원리금_계산(10000000, 0.05, 5)
print(f"2차 원리금: {원리금_2}원")
```

### **2) 함수 정의와 호출**
- **함수를 정의하는 것은 마치 레시피를 작성하는 것과 같음.**
    - 레시피에 필요한 재료와 순서가 적혀 있는 것처럼, 함수에는 어떤 작업을 할 것인지와 **필요한 값(매개변수)**이 포함됨.
    - 한 번 작성된 레시피는 여러 번 재사용 가능하듯, 함수도 한 번 정의되면 필요할 때마다 호출해서 사용할 수 있음.
- 함수를 정의할 때는 def 키워드를 사용함



- def: 정의를 시작할 때 사용하는 키워드.
- 함수이름: 호출할 때 사용할 이름. 이름은 코드를 이해하기 쉽게 작성하는 것이 좋음.
- 매개변수: 외부로부터 받는 값. 여러 개의 매개변수를 정의할 수 있음.
- 수행할 코드 : 실제로 동작하는 것을 코드로 작성, 레시피
- return: 함수가 결과 값을 반환할 때 사용하는 키워드. 반환값이 없을 수도 있음

```python
'''def 함수이름(매개변수1, 매개변수2):
    # 함수에서 수행할 코드
    return 반환값'''
```

## **함수 반환 값**

- 반환 값은 함수가 작업을 완료한 후 결과로 돌려주는 값임.
    - 마치 음식 배달처럼, 함수를 통해 요청한 작업의 결과가 집으로 배달된다고 생각할 수 있음.
    - 함수를 호출하고 나면, 그 결과를 다시 받아서 다른 작업에 사용할 수 있음

#### **함수의 반환 값은 마치 자판기와 같음.**
- 돈(매개변수)을 넣고 버튼(함수)을 누르면, 원하는 음료수(반환 값)가 나오는 것처럼, 함수는 필요한 값을 받아 작업을 수행한 후 결과를 돌려줌.

- 반환 값을 통해 함수가 작업한 결과를 다른 코드에서 활용할 수 있음

#### **문제 1 : 이익률 계산 함수**
- 매출액과 비용을 입력받아 이익률(이익/매출)을 계산하여 반환

```python
def 이익률(매출액, 비용):
    이익 = 매출액 - 비용
    이익률 = (이익 / 매출액) * 100  # 백분율로 계산
    return 이익률
```

#### **문제 2: 월간 평균 매출 계산 함수**
- 월간 매출액을 리스트로 입력받아 그 평균을 계산하는 함수를 작성함
- 월간 매출액 1000, 2000, 1500, 3000 만원

```python
def 평균매출(매출리스트):
    총합 = sum(매출리스트)
    평균 = 총합 / len(매출리스트)
    return 평균
```

```python
결과 = 평균매출([1000, 2000, 1500, 3000])
print(f"평균 매출액: {결과}원")
```

#### **문제 3: 특정 주식 보유 여부 함수 만들기**
- 보유 중인 주식 리스트가 주어졌을 때, 특정 주식이 리스트에 포함되어 있는지 확인하는 보유_확인 함수를 정의하세요.
- 이 함수는 주식 이름을 입력받아, 리스트에 포함되어 있으면 **"보유 중"** 을 반환하고, 포함되어 있지 않으면 **"보유 중 아님"** 을 반환함
- 보유_주식 = ["삼성전자", "카카오", "네이버", "LG화학"]

```python
보유_주식 = ["삼성전자", "카카오", "네이버", "LG화학"]

def 보유_확인(주식):
    if 주식 in 보유_주식:
        return "보유 중"
    else:
        return "보유 중 아님"

print(보유_확인("카카오"))  # "보유 중" 출력
print(보유_확인("현대차"))  # "보유 중 아님" 출력
```

#### **문제 4: VIP 고객 확인 함수 만들기**
- 은행의 VIP 고객 리스트가 주어져 있을 때, 고객 이름을 입력받아 VIP 여부를 확인하는 VIP_확인 함수를 정의하세요.
- 이 함수는 고객 이름을 받아, VIP 고객이면 "VIP 고객임"을 반환하고, 일반 고객이면 "일반 고객임"을 반환함.
- VIP_고객 = ["김철수", "이영희", "박민수", "최지현"]

```python
VIP_고객 = ["김철수", "이영희", "박민수", "최지현"]

def VIP_확인(고객):
    if 고객 in VIP_고객:
        return "VIP 고객임"
    else:
        return "일반 고객임"

print(VIP_확인("이영희"))  # "VIP 고객임" 출력
print(VIP_확인("박영수"))  # "일반 고객임" 출력
```

#### **문제 5: 재고 확인 함수 만들기**
- 물품과 각 물품의 재고량이 딕셔너리로 주어졌을 때, 특정 물품의 재고가 있는지 확인하는 재고_확인 함수를 정의하세요.
- 이 함수는 물품 이름을 입력받아 재고가 있으면 "재고 있음"을, 없으면 "재고 부족"을 반환함.
- 재고 = {"우유": 10, "빵": 0, "계란": 5, "사과": 2}

```python
재고 = {"우유": 10, "빵": 0, "계란": 5, "사과": 2}

def 재고_확인(물품):
    if 물품 in 재고 and 재고[물품] > 0:
        return "재고 있음"
    else:
        return "재고 부족"

print(재고_확인("우유"))  # "재고 있음" 출력
print(재고_확인("빵"))   # "재고 부족" 출력
```

## **매개변수와 인자**

- **매개변수(parameter)**: 함수가 값을 받을 준비를 하는 변수임.
- **인자(argument)**: 함수를 호출할 때 실제로 전달하는 값임.

#### **매개변수는 배달 앱의 주문 입력란, 인자는 그 입력란에 입력하는 실제 주문임 (예: 피자 1판, 콜라 1병)**

#### **1) 매개변수의 기본 동작**
-  매개변수는 필요한 만큼 정의할 수 있으며, 함수 호출 시 인자를 순서에 맞게 전달해야 함.
- 기본적으로 매개변수는 위치 인자로 사용되며, 함수 호출 시 정해진 순서대로 값을 받음

```python
# 함수 정의
def 출력(순서1, 순서2):
    print(f"첫번째 인자 {순서1}, 두번째 인자 {순서2}")

# 함수 호출: 인자 순서에 맞게 전달
출력("1번", "2번")
출력("2번", "1번")
```

```python
# 함수 정의: 이름과 나이를 출력하는 함수
def 자기소개(이름, 나이):
    print(f"안녕하세요, 제 이름은 {이름}이고, 나이는 {나이}살입니다.")

# 함수 호출: 인자 순서에 맞게 전달
자기소개("홍길동", 25)
자기소개(25, "홍길동")
```

#### **2) 복수의 매개변수 사용 (가변 인자)**

**A. 가변 인자**(*args)
- 가변 인자는 매개변수의 개수를 동적으로 정의할 수 있는 방식임.
- 매개변수 앞에 *을 붙여서 여러 개의 값을 한꺼번에 받을 수 있음.
- *args 여러 개의 인자를 튜플로 묶어서 함수 내부로 전달함.

```python
# 함수 정의: 여러 숫자를 더하는 함수
def 더하기(*args):
    결과 = 0
    for 숫자 in args:
        결과 += 숫자
    return 결과
```

```python
# 함수 호출 : 가변 인자를 전달
합 = 더하기(10, 20, 30, 40)
print(f"결과 : {합}")
```

**B. 키워드 가변 인자** (**kwargs)
- 키워드 가변 인자는 함수가 이름이 붙은 인자를 여러 개 받을 수 있게 해줌.
- 매개변수 앞에 **를 붙여서, 키워드로 전달된 인자들을 딕셔너리 형태로 받을 수 있음

```python
# 함수 정의: 키워드 가변 인자를 사용하는 함수
def 금융_상품_정보(**kwargs):
    for 키, 값 in kwargs.items():
        print(f"{키}: {값}")

# 함수 호출: 키워드 인자를 전달
금융_상품_정보(종목="삼성전자", 가격=80000, 수익률=5)
```

**C.기본값 매개변수**
- 매개변수에 기본값을 미리 설정할 수 있음.
- 함수 호출 시 인자를 전달하지 않으면 기본값을 사용하고, 인자를 전달하면 전달된 값을 사용함.

```python
# 함수 정의: 기본값을 가지는 매개변수
def 환영_인사(이름, 인사말="안녕하세요"):
    print(f"{인사말}, {이름}님!")

# 기본값을 사용하는 호출
환영_인사("홍길동")  # 출력: 안녕하세요, 홍길동님!

# 기본값을 무시하고 새로운 값을 전달한 호출
환영_인사("이순신", "환영합니다")
```

#### **3) 매개변수 순서**
- 가변 인자와 기본값 매개변수를 함께 사용할 때는 순서가 중요함.
- 일반 매개변수 → 기본값 매개변수 → 가변 인자 → 키워드 가변 인자의 순서로 정의해야 함.

```python
# 함수 정의: 매개변수 순서 중요
def 정보_출력(이름, 나이=20, *취미, **기타):
    print(f"이름: {이름}, 나이: {나이}")
    print(f"취미: {취미}")
    print(f"기타: {기타}")

# 함수 호출
정보_출력("홍길동", 25, "독서", "축구", 직업="개발자", 국적="한국")
```

## **지역변수 전역변수**

#### **1. 지역 변수란?**
- 지역 변수는 함수 내에서 정의된 변수로, 함수 내부에서만 유효함.
- 함수가 실행될 때 생성되고, 함수가 종료되면 사라짐.
- **지역 변수는 사무실 안에서만 쓸 수 있는 기기와 같음. 사무실 밖에서는 그 기기를 사용할 수 없음. 즉, 함수 밖에서는 그 변수를 사용할 수 없음**

```python
# 함수 정의
def 함수():
    숫자 = 10  # 지역 변수
    print(f"함수 내부에서: {숫자}")
```

```python
# 함수 호출
함수()  # 출력: 함수 내부에서: 10

# 함수 외부에서 지역 변수에 접근 시도
print(숫자)
```

#### **2. 전역 변수**
- 전역 변수는 함수 외부에서 정의된 변수로, 프로그램 전체에서 접근할 수 있음.
- 프로그램이 실행되는 동안 유효하며, 모든 함수에서 접근 가능함.
- 공용 냉장고와 같음.모든 사람들이 그 냉장고에서 물건을 꺼낼 수 있는 것처럼, 전역 변수는 모든 함수에서 접근 가능함.

```python
# 전역 변수 정의
메시지 = "안녕하세요!"

# 함수 정의
def 인사():
    print(메시지)

# 함수 호출
인사()
```

```python
메시지 = "안녕하세요!"

# 함수 정의
def 인사():
    숫자 = 10  # 지역 변수
    print(메시지)  # 전역 변수에 접근
    print(숫자)  # 지역 변수에 접근
```

```python
# 함수 호출
인사()
```

```python
# 함수 외부에서 지역 변수 접근 불가
print(메시지)
print(숫자)
```

#### **global 변수**
- 함수 내부에서 전역 변수를 수정할 때는 global 키워드를 사용해야 함.
- 전역 변수는 기본적으로 함수 내부에서 읽기 전용이므로, 수정하려면 명시적으로 선언해야 

```python
# 전역 변수 정의
카운터 = 0

# 함수 정의
def 증가():
    global 카운터
    카운터 += 1

# 함수 호출
증가()
증가()

print(카운터)
```

#### **문제1) 주식 수익률 계산 함수**
- 특정 주식의 매입가와 매도가를 입력받아, 수익률을 계산하는 함수를 작성하세요.
- 수익률 공식: 수익률(%) = ((매도가 - 매입가) / 매입가) * 100
- 사용자가 특정 주식을 매입한 가격과 매도한 가격을 입력했을 때, 그 주식의 수익률을 계산해야 함

```python
# 함수 정의
def 주식_수익률_계산(매입가, 매도가):
    수익률 = ((매도가 - 매입가) / 매입가) * 100  # 수익률 계산
    return 수익률
```

```python
# 함수 호출
결과 = 주식_수익률_계산(50000, 60000)  # 매입가 50,000원, 매도가 60,000원
print(f"주식 수익률: {결과}%")
```

#### **문제 2: 학급 평균 점수 계산 함수**
- 학생들의 이름, 점수를 입력받아 학급 전체의 **평균 점수**와 **최고 점수**, **최저 점수**를 계산하는 함수를 작성하세요.
- 학급 평균 점수는 모든 학생의 점수를 더한 후 학생 수로 나누어 계산함

```python
# 함수 정의
def 학급_점수_계산(**학생정보):
    총점 = 0
    최고점 = max(학생정보.values())
    최저점 = min(학생정보.values())
    학생수 = len(학생정보)
    for 점수 in 학생정보.values():
        총점 += 점수
    평균점수 = 총점 / 학생수
    return 평균점수, 최고점, 최저점
```

```python
# 함수 호출
평균점수, 최고점, 최저점 = 학급_점수_계산(홍길동=90, 이순신=85, 강감찬=95, 유관순=80)
print(f"평균 점수: {평균점수}, 최고 점수: {최고점}, 최저 점수: {최저점}")
```

#### **문제 4: 고가 주식 찾기**
- 주식 가격 정보가 딕셔너리로 주어졌을 때, 일정 금액 이상인 주식들을 리스트로 반환하는 고가_주식 함수를 작성하세요
- 주식_가격 = {"삼성전자": 80000, "카카오": 110000, "네이버": 120000, "LG화학": 85000}

```python
주식_가격 = {"삼성전자": 80000, "카카오": 110000, "네이버": 120000, "LG화학": 85000}

def 고가_주식(가격정보, 기준가격):
    결과 = []
    for 주식, 가격 in 가격정보.items():
        if 가격 >= 기준가격:
            결과.append(주식)
    return 결과

# 테스트
print(고가_주식(주식_가격, 90000))
print(고가_주식(주식_가격, 80000))
```

#### **문제 5: 목표 도달 시점 확인**
- 일정 금액 이상이 될 때까지 매월 투자하는 경우, 목표 금액에 도달하는 시점을 계산하는 목표_도달_시점 함수를 작성하세요.
- 매월 투자 금액과 연이율이 주어졌을 때 목표 금액에 도달하는 데 걸린 달 수를 반환함.

```python
def 목표_도달_시점(매월_투자, 이율, 목표금액):
    총금액 = 0
    달수 = 0
    월이율 = (이율 / 100) / 12
    while 총금액 < 목표금액:
        총금액 += 매월_투자
        총금액 += 총금액 * 월이율
        달수 += 1
    return 달수

# 테스트
print(목표_도달_시점(500000, 5, 10000000))
```
