### **1) 함수란**
- 동작을 미리 정의해두고, 필요할 때 그 동작을 반복해서 사용할 수 있게 해주는 도구임.
- 마치 레시피처럼, 함수를 한 번 만들어 놓으면 같은 작업을 여러 번 반복할 때마다 다시 적지 않고, 필요할 때 호출해서 사용할 수 있음.



#### **함수의 장점**
- **1. 코드의 재사용성**
    - 함수를 사용하면 한 번 정의한 동작을 여러 번 재사용할 수 있음.
    - 같은 기능을 하는 코드를 반복해서 작성할 필요가 없어짐.
    - 예)이자율 계산을 해야 한다고 가정하면, 매번 계산 공식을 적지 않고 함수로 한 번 정의해두고 필요할 때마다 그 함수를 불러오면 됨
- **2. 코드의 가독성 향상**
    - 코드가 더 짧고 명확해짐.
    - 코드를 읽을 때 함수 이름만 봐도 그 함수가 어떤 동작을 수행하는지 쉽게 알 수 있음.
- **3. 유지보수 용이성**
    - 코드의 중복을 줄이기 때문에 나중에 코드가 변경되더라도 한 번에 쉽게 수정할 수 있음.
    - 만약 함수에서 사용된 동작을 수정해야 한다면, 함수 정의 부분만 수정하면 됨.

#### **함수 없는 코드**
- 함수를 사용하지 않으면, 같은 계산을 여러 번 직접 작성해야 함.

```python
# 5년 동안 원금 1000만원, 연 이자율 5%를 적용해 원리금을 계산

# 첫번째 고객의 원금, 이자율, 기간
원금 = 10000000
이자율 = 0.05
기간 = 5

# 첫 번째 이자 계산
원리금_1 = 원금 + (원금 * 이자율 * 기간)
print(f"1차 원리금: {원리금_1}원")
```

```python
# 두번째 고객
원금 = 20000000
이자율 = 0.03
기간 = 3

# 두 번째 이자 계산 (다시 작성해야 함)
원리금_2 = 원금 + (원금 * 이자율 * 기간)
print(f"2차 원리금: {원리금_2}원")
```

#### **함수 있는 코드**
- 같은 계산을 함수로 정의하면, 코드를 반복해서 작성할 필요가 없음. 함수만 한 번 호출하면 됨

```python
# 이자율 계산 함수 정의
def 원리금_계산(원금, 이자율, 기간):
    return 원금 + (원금 * 이자율 * 기간)

# 함수 호출로 원리금 계산
원리금_1 = 원리금_계산(10000000, 0.05, 5)
print(f"1차 원리금: {원리금_1}원")

원리금_2 = 원리금_계산(10000000, 0.05, 5)
print(f"2차 원리금: {원리금_2}원")
```

### **2) 함수 정의와 호출**
- **함수를 정의하는 것은 마치 레시피를 작성하는 것과 같음.**
    - 레시피에 필요한 재료와 순서가 적혀 있는 것처럼, 함수에는 어떤 작업을 할 것인지와 **필요한 값(매개변수)**이 포함됨.
    - 한 번 작성된 레시피는 여러 번 재사용 가능하듯, 함수도 한 번 정의되면 필요할 때마다 호출해서 사용할 수 있음.
- 함수를 정의할 때는 def 키워드를 사용함



- def: 정의를 시작할 때 사용하는 키워드.
- 함수이름: 호출할 때 사용할 이름. 이름은 코드를 이해하기 쉽게 작성하는 것이 좋음.
- 매개변수: 외부로부터 받는 값. 여러 개의 매개변수를 정의할 수 있음.
- 수행할 코드 : 실제로 동작하는 것을 코드로 작성, 레시피
- return: 함수가 결과 값을 반환할 때 사용하는 키워드. 반환값이 없을 수도 있음

```python
'''def 함수이름(매개변수1, 매개변수2):
    # 함수에서 수행할 코드
    return 반환값'''
```

#### **함수 호출**
- 함수를 호출하는 것은 마치 레시피대로 요리하는 것과 같음.
    - 한 번 작성한 레시피(함수)를 필요할 때 불러(호출) 사용함.
    - 함수가 정의된 후, 함수 이름과 함께 필요한 값을 넣어 호출하면, 그 함수가 실행됨

```python
# 함수 정의
def 인사말(이름):
    print(f"안녕하세요, {이름}님!")
```

```python
# 함수 호출
인사말("홍길동")
```

### **3) 매개변수,인자**
- 매개변수 : 함수가 외부에서 전달받는 값을 받기 위해 사용하는 변수임. 함수 정의 시 매개변수를 설정함
- 인자 : 함수가 호출될 때 전달하는 실제 값임. 함수를 호출할 때 매개변수에 인자 값을 넣어줌

```python
# 함수 정의: 매개변수로 두 숫자를 받아 더하는 함수
def 더하기(숫자1, 숫자2):
    print(숫자1 + 숫자2)
```

```python
# 함수 호출: 인자로 값을 전달
더하기(10, 20)
```

- 함수 **더하기(숫자1, 숫자2)**는 두 개의 값을 더하는 함수임.
- 호출할 때 인자로 10과 20을 전달하여, 함수 내부의 매개변수 **숫자1, 숫자2**에 각각 할당됨.

#### **매개변수가 없는 함수**
- 함수는 매개변수를 사용하지 않고도 정의할 수 있음.
- 외부에서 값을 받지 않고 고정된 동작만 수행함

```python
# 매개변수가 없는 함수 정의
def 환영인사():
    print("환영합니다!")
```

```python
환영인사()
```

#### **3. 문제**
- 두 개의 숫자를 입력받아 더 큰 값을 출력하는 함수를 작성
- 두 숫자가 같다면 "두 숫자는 같습니다."라는 메시지를 출력

```python
# 함수 정의
def 큰값_출력(숫자1, 숫자2):
    if 숫자1 > 숫자2:
        print(숫자1)
    elif 숫자2 > 숫자1:
        print(숫자2)
    else:
        print("두 숫자는 같습니다.")
```

```python
큰값_출력(10, 20)
```
