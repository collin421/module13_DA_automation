#### **문제1) 주식 수익률 계산 함수**
- 특정 주식의 매입가와 매도가를 입력받아, 수익률을 계산하는 함수를 작성하세요.
- 수익률 공식: 수익률(%) = ((매도가 - 매입가) / 매입가) * 100
- 사용자가 특정 주식을 매입한 가격과 매도한 가격을 입력했을 때, 그 주식의 수익률을 계산해야 함

```python
# 함수 정의
def 주식_수익률_계산(매입가, 매도가):
    수익률 = ((매도가 - 매입가) / 매입가) * 100  # 수익률 계산
    return 수익률
```

```python
# 함수 호출
결과 = 주식_수익률_계산(50000, 60000)  # 매입가 50,000원, 매도가 60,000원
print(f"주식 수익률: {결과}%")
```

#### **문제 2: 학급 평균 점수 계산 함수**
- 학생들의 이름, 점수를 입력받아 학급 전체의 **평균 점수**와 **최고 점수**, **최저 점수**를 계산하는 함수를 작성하세요.
- 학급 평균 점수는 모든 학생의 점수를 더한 후 학생 수로 나누어 계산함

```python
# 함수 정의
def 학급_점수_계산(**학생정보):
    총점 = 0
    최고점 = max(학생정보.values())
    최저점 = min(학생정보.values())
    학생수 = len(학생정보)
    for 점수 in 학생정보.values():
        총점 += 점수
    평균점수 = 총점 / 학생수
    return 평균점수, 최고점, 최저점
```

```python
# 함수 호출
평균점수, 최고점, 최저점 = 학급_점수_계산(홍길동=90, 이순신=85, 강감찬=95, 유관순=80)
print(f"평균 점수: {평균점수}, 최고 점수: {최고점}, 최저 점수: {최저점}")
```

#### **문제 3: 은행 계좌 잔액 조회**
- 은행 계좌의 잔액 정보를 딕셔너리로 저장하고, 특정 계좌 번호가 잔액 조회 리스트에 있는지 확인하는 잔액_조회 함수를 작성하세요. 
- 잔액이 있으면 금액을, 없으면 "계좌 없음"을 반환함.
- 계좌_정보 = {"123-456": 100000, "234-567": 250000, "345-678": 50000}

```python
계좌_정보 = {"123-456": 100000, "234-567": 250000, "345-678": 50000}

def 잔액_조회(계좌번호):
    if 계좌번호 in 계좌_정보:
        return f"잔액: {계좌_정보[계좌번호]}원"
    else:
        return "계좌 없음"

# 테스트
print(잔액_조회("123-456"))  # "잔액: 100000원" 출력
print(잔액_조회("111-222"))  # "계좌 없음" 출력
```

#### **문제 4: 고가 주식 찾기**
- 주식 가격 정보가 딕셔너리로 주어졌을 때, 일정 금액 이상인 주식들을 리스트로 반환하는 고가_주식 함수를 작성하세요
- 주식_가격 = {"삼성전자": 80000, "카카오": 110000, "네이버": 120000, "LG화학": 85000}

```python
주식_가격 = {"삼성전자": 80000, "카카오": 110000, "네이버": 120000, "LG화학": 85000}

def 고가_주식(가격정보, 기준가격):
    결과 = []
    for 주식, 가격 in 가격정보.items():
        if 가격 >= 기준가격:
            결과.append(주식)
    return 결과

# 테스트
print(고가_주식(주식_가격, 90000))
print(고가_주식(주식_가격, 80000))
```

#### **문제 5: 목표 도달 시점 확인**
- 일정 금액 이상이 될 때까지 매월 투자하는 경우, 목표 금액에 도달하는 시점을 계산하는 목표_도달_시점 함수를 작성하세요.
- 매월 투자 금액과 연이율이 주어졌을 때 목표 금액에 도달하는 데 걸린 달 수를 반환함.

```python
def 목표_도달_시점(매월_투자, 이율, 목표금액):
    총금액 = 0
    달수 = 0
    월이율 = (이율 / 100) / 12
    while 총금액 < 목표금액:
        총금액 += 매월_투자
        총금액 += 총금액 * 월이율
        달수 += 1
    return 달수

# 테스트
print(목표_도달_시점(500000, 5, 10000000))
```
