## **함수 반환 값**

- 반환 값은 함수가 작업을 완료한 후 결과로 돌려주는 값임.
    - 마치 음식 배달처럼, 함수를 통해 요청한 작업의 결과가 집으로 배달된다고 생각할 수 있음.
    - 함수를 호출하고 나면, 그 결과를 다시 받아서 다른 작업에 사용할 수 있음

#### **함수의 반환 값은 마치 자판기와 같음.**
- 돈(매개변수)을 넣고 버튼(함수)을 누르면, 원하는 음료수(반환 값)가 나오는 것처럼, 함수는 필요한 값을 받아 작업을 수행한 후 결과를 돌려줌.



## **1) 반환 값이 없는 함수**
- 반환 값이 없는 함수는 **return**을 생략하거나, **return None**을 사용할 수 있음.
- 반환 값이 없을 때는 함수가 그저 동작만 하고 결과를 돌려주지 않음.

#### **매게변수 없이, 반환값 없이, 코카콜라를 출력하는 자판기 함수**

```python
# 반환 값이 없는 함수 정의
def 자판기():
    print("코카콜라")  # 화면에 출력만 하고 반환 값은 없음

# 함수 호출
자판기()
```

```python
음료 = 자판기()
print(음료)
```

## **2) 반환 값이 있는 함수**
- 함수에서 **return 키워드를** 사용하여 결과를 반환함.
- 반환 값을 통해 함수가 작업한 결과를 다른 코드에서 활용할 수 있음

```python
# 반환 값이 없는 함수 정의
def 자판기():
    print("코카콜라")  # 화면에 출력만 하고 반환 값은 없음
    음료 = "코카콜라"
    return 음료  # 반환 값

# 함수 호출
자판기()
```

```python
음료 = 자판기()
print(음료)
```

## **3) 여러 값을 반환하는 함수**
- 파이썬에서는 함수가 여러 개의 값을 동시에 반환할 수 있음.
    - 함수에서 여러 값을 튜플로 묶어 반환하고, 튜플 언패킹을 통해 각 값을 받아올 수 있음

```python
# 반환 값이 없는 함수 정의
def 자판기():
    음료1 = "코카콜라"
    음료2 = "사이다"
    음료3 = "환타"
    print(음료1, 음료2, 음료3)
    # return(음료1, 음료2, 음료3) # 여러개의 반환값 

# 함수 호출
자판기()
```

```python
# 함수 정의: 두 숫자의 합과 차를 반환하는 함수
def 합과_차(숫자1, 숫자2):
    합계 = 숫자1 + 숫자2
    차이 = 숫자1 - 숫자2
    return 합계, 차이  # 두 값을 동시에 반환

# 함수 호출: 여러 값 받기
합, 차 = 합과_차(20, 10)
print(f"합계: {합}, 차이: {차}")
```

## **연습 문제**

#### **문제 1 : 이익률 계산 함수**
- 매출액과 비용을 입력받아 이익률(이익/매출)을 계산하여 반환

```python
def 이익률(매출액, 비용):
    이익 = 매출액 - 비용
    이익률 = (이익 / 매출액) * 100  # 백분율로 계산
    return 이익률
```

```python
결과 = 이익률(5000, 3000)
print(f"이익률: {결과}%")
```

#### **문제 2: 월간 평균 매출 계산 함수**
- 월간 매출액을 리스트로 입력받아 그 평균을 계산하는 함수를 작성함
- 월간 매출액 1000, 2000, 1500, 3000 만원

```python
def 평균매출(매출리스트):
    총합 = sum(매출리스트)
    평균 = 총합 / len(매출리스트)
    return 평균
```

```python
결과 = 평균매출([1000, 2000, 1500, 3000])
print(f"평균 매출액: {결과}원")
```

#### **문제 3: 특정 주식 보유 여부 함수 만들기**
- 보유 중인 주식 리스트가 주어졌을 때, 특정 주식이 리스트에 포함되어 있는지 확인하는 보유_확인 함수를 정의하세요.
- 이 함수는 주식 이름을 입력받아, 리스트에 포함되어 있으면 **"보유 중"** 을 반환하고, 포함되어 있지 않으면 **"보유 중 아님"** 을 반환함
- 보유_주식 = ["삼성전자", "카카오", "네이버", "LG화학"]

```python
보유_주식 = ["삼성전자", "카카오", "네이버", "LG화학"]

def 보유_확인(주식):
    if 주식 in 보유_주식:
        return "보유 중"
    else:
        return "보유 중 아님"

print(보유_확인("카카오"))  # "보유 중" 출력
print(보유_확인("현대차"))  # "보유 중 아님" 출력
```

#### **문제 4: VIP 고객 확인 함수 만들기**
- 은행의 VIP 고객 리스트가 주어져 있을 때, 고객 이름을 입력받아 VIP 여부를 확인하는 VIP_확인 함수를 정의하세요.
- 이 함수는 고객 이름을 받아, VIP 고객이면 "VIP 고객임"을 반환하고, 일반 고객이면 "일반 고객임"을 반환함.
- VIP_고객 = ["김철수", "이영희", "박민수", "최지현"]

```python
VIP_고객 = ["김철수", "이영희", "박민수", "최지현"]

def VIP_확인(고객):
    if 고객 in VIP_고객:
        return "VIP 고객임"
    else:
        return "일반 고객임"

print(VIP_확인("이영희"))  # "VIP 고객임" 출력
print(VIP_확인("박영수"))  # "일반 고객임" 출력
```

#### **문제 5: 재고 확인 함수 만들기**
- 물품과 각 물품의 재고량이 딕셔너리로 주어졌을 때, 특정 물품의 재고가 있는지 확인하는 재고_확인 함수를 정의하세요.
- 이 함수는 물품 이름을 입력받아 재고가 있으면 "재고 있음"을, 없으면 "재고 부족"을 반환함.
- 재고 = {"우유": 10, "빵": 0, "계란": 5, "사과": 2}

```python
재고 = {"우유": 10, "빵": 0, "계란": 5, "사과": 2}

def 재고_확인(물품):
    if 물품 in 재고 and 재고[물품] > 0:
        return "재고 있음"
    else:
        return "재고 부족"

print(재고_확인("우유"))  # "재고 있음" 출력
print(재고_확인("빵"))   # "재고 부족" 출력
```
