## 1. 파이썬_클래스01

### **1. 클래스란 무엇인가?**
- 클래스는 **객체 지향 프로그래밍(OOP)**의 기본 개념으로, 데이터와 기능을 묶어 특정 목적의 객체를 생성할 수 있는 설계도임.
- 파이썬에서 클래스를 사용하면 속성(데이터)과 메서드(기능)을 포함한 객체를 생성할 수 있어, 코드의 재사용성을 높이고 유지보수가 편리해짐.

### **2. 절차 지향 프로그래밍 vs. 객체 지향 프로그래밍**
**절차 지향 프로그래밍 (Procedural Programming)**
- 함수와 명령어의 순서에 따라 프로그램을 실행하며 데이터와 기능이 분리되어 있음.
- 작은 프로그램에는 효과적이나, 복잡한 시스템에서 코드 중복, 유지보수 어려움 등의 문제가 발생함.
</br>

**객체 지향 프로그래밍 (OOP)**
- 데이터와 기능을 객체로 묶어 클래스로 정의하고, 이를 통해 개별 객체를 생성하여 기능을 구현함.
- 데이터와 메서드를 캡슐화하여 외부 접근을 제한하고, 상속을 통해 코드 재사용성을 높임.

**절차 지향에서 객체 지향으로 전환된 이유**
- 절차 지향 프로그래밍은 데이터와 함수가 분리되어 있고 코드 중복이 많아 유지보수와 확장성이 부족한 한계를 가짐.
- 이에 따라 객체 지향 프로그래밍이 등장하여, 데이터와 기능을 객체로 묶어 캡슐화하고, 상속과 다형성으로 재사용성과 확장성을 높이며 현실적 모델링이 가능해짐.


### **3. 객체 지향 프로그래밍의 장점**
- **코드 재사용성:** 상속을 통해 기존 코드를 재사용하고, 중복을 줄일 수 있음.
- **유지보수 용이성:** 코드가 모듈화되어 있어 수정이 용이하고 협업에 유리함.
- **확장성:** 다형성으로 같은 메서드가 클래스에 따라 다르게 동작하여 확장성이 높음.
- **데이터 보안:** 캡슐화를 통해 객체 내부 데이터를 보호하고, 외부에서 직접 변경하지 못하도록 함.
- **현실적 모델링:** 객체 지향은 현실 세계 개념을 코드로 표현하기 쉬워, 복잡한 문제를 직관적으로 해결 가능함.

#### **문제 2: 동물 클래스 만들기**
- Animal이라는 클래스를 정의하세요.
- Animal 클래스는 species (종류)와 age (나이) 속성을 가집니다.
- __init__ 생성자를 사용하여 species는 "개", age는 5로 초기화하도록 하세요.
- sound()라는 메서드를 추가하여 "동물이 소리를 냅니다."를 출력하게 하세요

```python
class Animal:
    def __init__(self):
        self.species = "개"  # 종류 속성 초기값으로 "개" 설정
        self.age = 5         # 나이 속성 초기값으로 5 설정

    def sound(self):
        print("동물이 소리를 냅니다.")  # 소리 메서드
```

#### **문제 3: 자동차 클래스와 메서드 작성**
- Car라는 클래스를 정의하세요.
- Car 클래스는 brand (브랜드), model (모델), speed (속도) 속성을 가집니다.
- __init__ 생성자를 통해 brand, model, speed 속성을 초기화할 수 있도록 하세요.
- accelerate(amount) 메서드를 정의하여 속도를 amount만큼 증가시키세요.
- brake(amount) 메서드를 정의하여 속도를 amount만큼 줄이세요. 단, 속도는 0 이하로 내려가지 않도록 설정하세요.

```python
class Car:
    def __init__(self, brand, model, speed=0):
        self.brand = brand
        self.model = model
        self.speed = speed

    def accelerate(self, amount):
        self.speed += amount
        print(f"속도가 {amount}만큼 증가하여 현재 속도: {self.speed} km/h")

    def brake(self, amount):
        self.speed = max(0, self.speed - amount)
        print(f"속도가 {amount}만큼 감소하여 현재 속도: {self.speed} km/h")
```

```python
# 객체 생성 및 메서드 호출
car = Car("현대", "소나타")
car.accelerate(50)  # 출력: 속도가 50만큼 증가하여 현재 속도: 50 km/h
car.brake(20)       # 출력: 속도가 20만큼 감소하여 현재 속도: 30 km/h
car.brake(40)       # 출력: 속도가 40만큼 감소하여 현재 속도: 0 km/h
```

# 파이썬 클래스 04 — 상속

생성자(__init__)에 이어 **상속(Inheritance)** 을 다룹니다. 부모 클래스의 기능을 물려받고, super()와 메서드 오버라이딩으로 확장합니다.

### **8. 상속 (Inheritance) — 기능 물려받기**

- **상속**이란 이미 만들어 둔 클래스(부모·슈퍼 클래스)의 속성과 메서드를 새 클래스(자식·서브 클래스)가 그대로 **물려받는** 것임.
- 형태: `class 자식클래스(부모클래스):`
- **왜 쓰나?** 같은 코드를 또 쓰지 않고(중복 제거) 부모 기능을 재사용하면서, 자식만의 기능을 더할 수 있음.

```python
# 부모(슈퍼) 클래스
class 동물:
    def __init__(self, 이름):
        self.이름 = 이름

    def 숨쉬기(self):
        print(f"{self.이름}: 숨을 쉽니다")
```

```python
# 자식(서브) 클래스 — 동물을 상속받음
class 강아지(동물):
    def 짖기(self):
        print(f"{self.이름}: 멍멍!")
```

```python
초코 = 강아지("초코")
초코.숨쉬기()   # 부모(동물)에게 물려받은 메서드
초코.짖기()     # 자식(강아지) 고유 메서드
```

#### **메서드 오버라이딩 (재정의)**
- 부모가 가진 메서드를 자식에서 **같은 이름으로 다시 정의**하면, 자식 객체는 자식 것을 우선 사용함.
- 부모의 기본 동작을 자식 상황에 맞게 바꿀 때 사용함.

```python
class 고양이(동물):
    def 숨쉬기(self):   # 부모의 숨쉬기를 덮어씀
        print(f"{self.이름}: 조용히 숨을 쉽니다 (냐옹)")

나비 = 고양이("나비")
나비.숨쉬기()   # 자식 버전이 호출됨
```

#### **super() — 부모 기능 가져다 쓰기**
- 자식 생성자(`__init__`)에서 부모의 `__init__`을 호출해 **부모 초기화를 재사용**함.
- `super().__init__(...)` 형태로 씀. 부모가 하던 일을 반복해서 적지 않아도 됨.

```python
class 동물:
    def __init__(self, 이름):
        self.이름 = 이름

class 강아지(동물):
    def __init__(self, 이름, 견종):
        super().__init__(이름)   # 부모 생성자 호출 → self.이름 설정
        self.견종 = 견종         # 자식만의 속성 추가

초코 = 강아지("초코", "푸들")
print(초코.이름, 초코.견종)
```

#### **상속 관계 확인 — isinstance / issubclass**
- `isinstance(객체, 클래스)`: 객체가 그 클래스(또는 부모 클래스)의 인스턴스인지
- `issubclass(자식, 부모)`: 자식이 부모를 상속했는지

```python
print(isinstance(초코, 강아지))  # True
print(isinstance(초코, 동물))    # True — 부모 타입으로도 인정됨
print(issubclass(강아지, 동물))  # True
```

### **연습문제 1. 사람 → 학생 상속**
- `사람` 클래스: 생성자로 (이름, 나이) 받고 `소개()` 메서드로 출력
- `학생` 클래스: `사람`을 상속, 생성자에서 `super()`로 이름·나이를 받고 **학번**을 추가
- `학생`의 `소개()`는 부모 소개를 부른 뒤 학번도 함께 출력 (오버라이딩 + super)

```python
class 사람:
    def __init__(self, 이름, 나이):
        self.이름 = 이름
        self.나이 = 나이
    def 소개(self):
        print(f"이름: {self.이름}, 나이: {self.나이}")

class 학생(사람):
    def __init__(self, 이름, 나이, 학번):
        super().__init__(이름, 나이)   # 부모 초기화 재사용
        self.학번 = 학번
    def 소개(self):                     # 오버라이딩
        super().소개()                  # 부모 소개 먼저
        print(f"학번: {self.학번}")

s = 학생("김파이", 20, "2026001")
s.소개()
```

#### **문제 2: 도형 클래스 상속 (area 오버라이딩)**
- 부모 `Shape`: `name` 속성, `area()`는 0을 반환
- `Rectangle(Shape)`: width·height → `area()` 재정의
- `Circle(Shape)`: radius → `area()` 재정의
- 여러 도형을 리스트에 담아 반복하며 넓이를 출력

```python
class Shape:
    def __init__(self, name):
        self.name = name
    def area(self):
        return 0

class Rectangle(Shape):
    def __init__(self, width, height):
        super().__init__("사각형")
        self.width = width
        self.height = height
    def area(self):
        return self.width * self.height

class Circle(Shape):
    def __init__(self, radius):
        super().__init__("원")
        self.radius = radius
    def area(self):
        return 3.14 * self.radius ** 2

for s in [Rectangle(3, 4), Circle(5)]:
    print(f"{s.name} 넓이: {s.area()}")
```

#### **문제 3 (도전): 동물 소리 — 상속 + 오버라이딩**
- 부모 `Animal`: 생성자로 `species`를 받고, `sound()`는 `"..."` 출력
- `Dog`, `Cat`: 각각 `Animal`을 상속하고 `sound()`를 재정의(멍멍 / 야옹)
- 동물 리스트를 돌며 각자 소리를 내게 함 (같은 호출, 다른 동작)

```python
class Animal:
    def __init__(self, species):
        self.species = species
    def sound(self):
        print("...")

class Dog(Animal):
    def __init__(self):
        super().__init__("개")
    def sound(self):
        print("멍멍")

class Cat(Animal):
    def __init__(self):
        super().__init__("고양이")
    def sound(self):
        print("야옹")

for a in [Dog(), Cat()]:
    print(a.species, end=": ")
    a.sound()
```
