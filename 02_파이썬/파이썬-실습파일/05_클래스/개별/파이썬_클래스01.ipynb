### **1. 클래스란 무엇인가?**
- 클래스는 **객체 지향 프로그래밍(OOP)**의 기본 개념으로, 데이터와 기능을 묶어 특정 목적의 객체를 생성할 수 있는 설계도임.
- 파이썬에서 클래스를 사용하면 속성(데이터)과 메서드(기능)을 포함한 객체를 생성할 수 있어, 코드의 재사용성을 높이고 유지보수가 편리해짐.

### **2. 절차 지향 프로그래밍 vs. 객체 지향 프로그래밍**
**절차 지향 프로그래밍 (Procedural Programming)**
- 함수와 명령어의 순서에 따라 프로그램을 실행하며 데이터와 기능이 분리되어 있음.
- 작은 프로그램에는 효과적이나, 복잡한 시스템에서 코드 중복, 유지보수 어려움 등의 문제가 발생함.
</br>

**객체 지향 프로그래밍 (OOP)**
- 데이터와 기능을 객체로 묶어 클래스로 정의하고, 이를 통해 개별 객체를 생성하여 기능을 구현함.
- 데이터와 메서드를 캡슐화하여 외부 접근을 제한하고, 상속을 통해 코드 재사용성을 높임.

**절차 지향에서 객체 지향으로 전환된 이유**
- 절차 지향 프로그래밍은 데이터와 함수가 분리되어 있고 코드 중복이 많아 유지보수와 확장성이 부족한 한계를 가짐.
- 이에 따라 객체 지향 프로그래밍이 등장하여, 데이터와 기능을 객체로 묶어 캡슐화하고, 상속과 다형성으로 재사용성과 확장성을 높이며 현실적 모델링이 가능해짐.


### **3. 객체 지향 프로그래밍의 장점**
- **코드 재사용성:** 상속을 통해 기존 코드를 재사용하고, 중복을 줄일 수 있음.
- **유지보수 용이성:** 코드가 모듈화되어 있어 수정이 용이하고 협업에 유리함.
- **확장성:** 다형성으로 같은 메서드가 클래스에 따라 다르게 동작하여 확장성이 높음.
- **데이터 보안:** 캡슐화를 통해 객체 내부 데이터를 보호하고, 외부에서 직접 변경하지 못하도록 함.
- **현실적 모델링:** 객체 지향은 현실 세계 개념을 코드로 표현하기 쉬워, 복잡한 문제를 직관적으로 해결 가능함.

#### **1) 클래스 (Class)**
- 클래스는 객체 지향 프로그래밍의 핵심 개념으로, 데이터와 기능을 하나로 묶어 정의한 설계도임.
- 클래스는 특정 개념을 표현하는 틀로, 여러 객체를 생성할 수 있는 청사진 역할을 함.

```python
class 붕어빵틀():
    pass
```

#### **2) 객체 (Object)**
- 객체는 클래스로부터 생성된 실체로, **클래스의 인스턴스(instance)** 라고도 불림.
- 각 객체는 클래스의 설계에 따라 속성과 메서드를 가질 수 있음.

```python
내빵 = 붕어빵틀()
너빵 = 붕어빵틀()
우리빵 = 붕어빵틀()
```

#### **3) 속성 (property)**
- 속성은 객체가 가지는 데이터를 나타내는 변수로, 객체의 상태나 특성을 정의함.
- 속성은 객체가 생성된 후에 추가할 수 있으며, 객체별로 개별적인 값을 가질 수 있음.

```python
내빵.앙꼬 = "단팥"
너빵.앙꼬 = "딸기맛"
우리빵.앙꼬 = "슈크림"
```

```python
print(내빵.앙꼬)
print(너빵.앙꼬)
print(우리빵.앙꼬)
```

#### **문제 1: 자동차 클래스**
- 자동차라는 이름의 클래스를 정의하세요.
- 첫 번째 자동차1이라는 객체를 만들고 브랜드 속성 값으로 **"현대"** 를, 두 번째 자동차2 객체에는 **"기아"** 를 저장하세요.
- 생성한 두 객체의 브랜드 속성 값을 화면에 출력하세요.

```python
class 자동차:
    pass
```

```python
자동차1 = 자동차()
자동차2 = 자동차()
```

```python
자동차1.브랜드 = '현대'
자동차2.브랜드 = '기아'
```

```python
print(자동차1.브랜드)
print(자동차2.브랜드)
```

#### **문제 2: 은행계좌 클래스**
- 은행계좌라는 이름의 클래스를 정의하고, 은행계좌 클래스로부터 두 개의 계좌 객체를 생성
- 다음표를 참조, 각 객체에 속성을 추가하고, 두 객체의 예금액과 이름, 연락처 속성을 저장하고
- 각 계좌의 속성을 모두 출력
    - 이름 : 박명수 / 예금액 : 1200 / 연락처 : 010-1234-5678
    - 이름 : 유재석 / 예금액 : 5000 / 연락처 : 010-4567-1234

```python
class 은행계좌:
    pass
```

```python
계좌1 = 은행계좌()
계좌2 = 은행계좌()
```

```python
계좌1.이름 = '박명수'
계좌1.예금액 = 1200
계좌1.연락처 = '010-4567-1234'
```

```python
계좌2.이름 = '유재석'
계좌2.예금액 = 5000
계좌2.연락처 = '010-4567-1234'
```

```python
print(f"계좌주 : {계좌1.이름}, 예금액 : {계좌1.예금액}, 연락처 : {계좌1.연락처}")
print(f"계좌주 : {계좌2.이름}, 예금액 : {계좌2.예금액}, 연락처 : {계좌2.연락처}")
```
