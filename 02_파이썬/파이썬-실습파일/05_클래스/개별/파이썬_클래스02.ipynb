### **4. 클래스 속성과 인스턴스 속성**
- **클래스 속성:** 클래스 자체에 정의되어 모든 객체가 공유하는 속성임.
- **인스턴스 속성:** 개별 객체에 정의된 속성으로, 각 객체가 독립적으로 소유함

```python
# 클래스 정의 하기
class 붕어빵틀:
    주재료 = '밀가루'
    pass
```

```python
내빵 = 붕어빵틀()
너빵 = 붕어빵틀()
```

```python
내빵.앙꼬 = '초콜릿'
너빵.앙꼬 = '딸기맛'
```

```python
print(내빵.앙꼬, 내빵.주재료)
print(너빵.앙꼬, 너빵.주재료)
```

### **5. 메서드 (Method)**
- 메서드는 클래스 내부에 정의된 함수로, 객체가 수행할 수 있는 기능을 나타냄.
- 메서드는 def 키워드로 정의되며, 클래스 내부에서 객체의 속성을 참조하거나 변경할 수 있음.

```python
class 붕어빵틀:
    def 앙꼬넣기(어떤빵, 넣을앙꼬):
        어떤빵.앙꼬 = 넣을앙꼬
```

```python
내빵 = 붕어빵틀()
너빵 = 붕어빵틀()
```

```python
붕어빵틀.앙꼬넣기(내빵, '딸기맛')
붕어빵틀.앙꼬넣기(너빵, '바나나맛')
```

```python
print(내빵.앙꼬)
print(너빵.앙꼬)
```

#### **2. 계좌 클래스** 
- 계좌 클래스 정의, 개설이라는 메서드를 정의하고, 개설 메서드에 누구계좌, 이름, 잔고 세개의 인자를 받음
- 개설 메서드는 객체 공강에 이름 잔고라는 속성을 저장하는 기능 만들기
- 김씨와 이씨의 계좌 객채를 만들고 이름과 잔고를 저장
    - 누구계좌 : 김씨 / 이름 : 김병수 / 잔고 : 65000
    - 누구계좌 : 박씨 / 이름 : 박수홍 / 잔고 : 125000

```python
class 계좌:
    def 개설(누구계좌, 이름, 잔고):
        누구계좌.이름 = 이름
        누구계좌.잔고 = 잔고
```

```python
김씨 = 계좌()
박씨 = 계좌()
```

```python
계좌.개설(김씨, '김병수', 65000)
계좌.개설(박씨, '박수홍', 65000)
```

```python
print(김씨.이름, 김씨.잔고)
print(박씨.이름, 박씨.잔고)
```

**3. 출력 메소드 만들기**
- 계좌 클래스에 출력이라는 메서드를 추가
- 출력이라는 메서드는 계좌 객체의 이름과 잔고 속성을 출력(프린트)하는 기능 정의
- 각 객체의 출력이라는 메서드 실행 하기

```python
class 계좌:
    def 개설(누구계좌, 이름, 잔고):
        누구계좌.이름 = 이름
        누구계좌.잔고 = 잔고
    def 출력(누구계좌):
        print(누구계좌.이름)
        print(누구계좌.잔고)
```

```python
계좌.출력(김씨)
```

```python
계좌.출력(박씨)
```
