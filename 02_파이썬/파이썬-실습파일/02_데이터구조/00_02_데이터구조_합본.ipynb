## **1. 리스트란 무엇인가?**

#### **리스트의 정의**
- 리스트는 여러 개의 데이터를 순차적으로 저장할 수 있는 파이썬의 데이터 구조임.
- 리스트는 대괄호 [] 안에 데이터를 **쉼표 ,**로 구분하여 저장함.
- 리스트는 다양한 데이터 타입을 동시에 저장할 수 있음.

#### **리스트 특징**
- 순서가 있음: 리스트의 각 요소는 인덱스를 가지고 있으며, 순서를 유지함.
- 변경 가능: 리스트의 요소는 생성 후에 변경할 수 있음

## **2. 리스트 생성 방법**

````python
# 주식종목 리스트
주식종목_리스트 = ['삼성전자', 'LG화학', '현대차', 'NAVER', '카카오', 'SK하이닉스', 'POSCO']
````

````python
# 금리 정보 리스트 (기준금리, 대출금리, 예금금리)
금리_리스트 = [1.25, 3.6, 2.1]
````

````python
# 혼합된 리스트 (주식종목, 시가총액)
혼합_리스트 = ['삼성전자', 350, '현대차', 50, 'NAVER', 40]
혼합_리스트2 = ['삼성전자', 350.12, '현대차', 50, 'NAVER', 400.0]
````

````python
print(주식종목_리스트)
print(금리_리스트)
print(혼합_리스트)
print(혼합_리스트2)
````

## **3. 리스트 요소 다루기**

**인덱싱과 슬라이싱**
- 인덱싱: 리스트의 각 요소는 인덱스 번호를 가지고 있으며, 이 번호를 통해 특정 요소에 접근할 수 있음. 인덱스는 0부터 시작함.
- 슬라이싱: 리스트의 일부 구간을 잘라내어 새로운 리스트로 반환할 수 있음

````python
주식종목_리스트 = ['삼성전자', 'LG화학', '현대차', 'NAVER', '카카오', 'SK하이닉스', 'POSCO']
````

````python
# 인덱싱
print(주식종목_리스트[1])
print(주식종목_리스트[-1])
````

````python
# 슬라이싱
print(주식종목_리스트[2:5])
print(주식종목_리스트[:3])
print(주식종목_리스트[4:])
````

````python
# index() 데이터의 인덱스 출력
주식종목_리스트.index("NAVER")
````

## **4. 리스트 요소 추가/삭제**

#### **추가**
- append(): 리스트의 끝에 새로운 요소를 추가함.
- insert(): 특정 위치에 새로운 요소를 삽입함.

````python
주식종목_리스트 = ['삼성전자', 'LG화학', '현대차']
````

````python
# 요소 추가 (끝에 추가)
주식종목_리스트.append('카카오')
print(주식종목_리스트)
````

````python
# 특정 위치에 요소 삽입
주식종목_리스트.insert(2, 'NAVER')  # 2번 인덱스에 'NAVER' 삽입
print(주식종목_리스트)
````

#### **삭제**
- del 리스트[인덱스] : 리스트의 특정 인덱스 제거
- remove(): 리스트에서 특정 값을 찾아 제거
- pop(): 리스트의 마지막 요소를 제거하고 반환. 특정 인덱스의 요소를 제거할 수도 있음

````python
주식종목_리스트 = ['삼성전자', 'LG화학', 'NAVER', '현대차', '카카오']
````

````python
# 특정 값 제거
주식종목_리스트.remove('NAVER')
print(주식종목_리스트)
````

````python
# 마지막 요소 제거
주식종목_리스트.pop()
print(주식종목_리스트)
````

````python
# 인덱스를 지정하여 요소 제거
주식종목_리스트.pop(2)  # 2번 인덱스 요소 제거
print(주식종목_리스트)
````

````python
# 인덱스를 지정하여 요소 제거2
del 주식종목_리스트[0]
print(주식종목_리스트)
````

#### **1. 딕셔너리란 무엇인가?**
- 딕셔너리 정의: 키와 값 쌍으로 데이터를 저장하는 파이썬의 데이터 구조.
- 딕셔너리는 대괄호 {} 안에 key: value 형식으로 데이터를 저장하고, 각 키는 고유하며 중복될 수 없음.
- 예시: {"삼성전자": 80000, "현대차": 200000, "카카오": 120000}

#### **2. 딕셔너리 생성 방법**
- 딕셔너리 기본 생성
- dict() 함수나 {}로 빈 딕셔너리를 생성하고 값을 추가

````python
주식_가격 = {}
주식_가격["삼성전자"] = 80000
주식_가격["현대차"] = 200000
주식_가격
````

#### **3. 딕셔너리 요소 다루기**
- 키로 값 접근하기 : 딕셔너리에서 키를 사용하여 특정 값을 가져올 수 있음.

````python
주식_가격 = {"삼성전자": 80000, "현대차": 200000, "카카오": 120000}
주식_가격["삼성전자"]
````

- 키와 값 추가 및 수정 : 딕셔너리에 새로운 키와 값을 추가하거나 기존 키의 값을 수정 가능.

````python
주식_가격["LG화학"] = 150000  # LG화학 추가
주식_가격["삼성전자"] = 82000  # 삼성전자 값 수정
````

#### **4. 딕셔너리 요소 삭제하기**
**요소 삭제 방법**
- del 키워드를 사용하여 특정 키-값 쌍 삭제.
- pop() 메서드를 사용하여 키로 값을 삭제하고 반환

````python
del 주식_가격["현대차"]  # 현대차 제거
주식_가격
````

````python
주식_가격.pop("카카오")  # 카카오 제거 및 값 반환
주식_가격
````

#### **5. 딕셔너리 유용한 메서드**
- keys(), values(), items() : 딕셔너리의 키, 값, 키-값 쌍을 각각 반환.

````python
print(주식_가격.keys())     # 모든 키 반환
print(주식_가격.values())   # 모든 값 반환
print(주식_가격.items())    # (키, 값) 쌍 반환
````

- get() : 키로 값을 가져오며, 키가 없을 경우 None 또는 지정한 기본값을 반환.

## 3. 데이터구조03_tuple_set

## **1. 튜플(Tuple)**
#### **튜플의 정의**
- 순서가 있는 자료형으로, 리스트와 유사하지만 **변경 불가능(Immutable)**함.
- 대괄호가 아닌 **소괄호 ()**를 사용하여 여러 개의 데이터를 저장함.
#### **튜플의 특징**
- 변경 불가: 한 번 정의된 튜플의 요소들은 수정, 삭제, 추가할 수 없음.
- 순서가 있음: 리스트처럼 인덱스를 통해 각 요소에 접근 가능.
#### **튜플의 장점**
- 안정성: 튜플은 수정할 수 없기 때문에, 데이터 보호가 필요한 경우 적합함.
- 메모리 효율: 리스트보다 메모리를 적게 사용함.

````python
# 주식 종목과 시가총액을 튜플로 저장
주식_튜플 = ('삼성전자', 350, '현대차', 50, 'LG화학', 45)

# 튜플 요소 접근
print(주식_튜플[0])
print(주식_튜플[1])
````

````python
# 금융 상품 코드 (변경되지 않는 정보)
금융_상품_코드 = ('KR7005930003', 'KR7005380001', 'KR7000270005')
금융_상품_코드
````

````python
# 주식 정보 튜플
주식_정보 = ('삼성전자', 350)

# 튜플 언패킹
종목명, 시가총액 = 주식_정보
print(종목명)
print(시가총액)
````

````python
# 튜플 생성
주식_튜플 = ('삼성전자', 'LG화학', '현대차')

# 튜플의 요소를 수정하려고 시도
주식_튜플[0] = '카카오'
````

````python
# 튜플 생성
주식_튜플 = ('삼성전자', 'LG화학', '현대차')

# 튜플을 리스트로 변환
주식_리스트 = list(주식_튜플)

# 리스트에서 수정
주식_리스트[0] = '카카오'

# 다시 튜플로 변환
주식_튜플 = tuple(주식_리스트)

print(주식_튜플)
````

## **2. 세트(Set)**
#### **세트의 정의**
- **세트(Set)**는 파이썬의 순서가 없는 자료형으로, 중복을 허용하지 않음.
- **중괄호 {}**를 사용하여 데이터 저장.
#### **세트의 특징**
- 중복 없음: 세트는 중복된 값을 허용하지 않음.
- 순서 없음: 세트는 요소의 순서가 보장되지 않음.
- 수학적 연산: 세트는 교집합, 합집합, 차집합 등 수학적 연산을 쉽게 수행할 수 있음.

````python
# 주식 종목 세트
주식_세트 = {'삼성전자', '현대차', 'LG화학', '삼성전자'}

print(주식_세트)
````

#### **교집합 (&) : 두 세트에 공통으로 포함된 요소를 반환함**

````python
주식_세트1 = {'삼성전자', 'LG화학', '현대차'}
주식_세트2 = {'카카오', '삼성전자', '현대차'}

교집합 = 주식_세트1 & 주식_세트2

print(교집합)
````

#### **합집합 (|) : 두 세트에 포함된 모든 요소를 반환하며, 중복은 제거됨.**

````python
합집합 = 주식_세트1 | 주식_세트2
print(합집합)
````

#### **차집합 (-) : 첫 번째 세트에는 있지만 두 번째 세트에는 없는 요소들을 반환함.**

````python
차집합 = 주식_세트1 - 주식_세트2
print(차집합)
````

#### **세트의 주요 메서드**
- add(): 세트에 새로운 요소를 추가함.
- remove(): 세트에서 특정 요소를 제거함.

````python
주식_세트 = {'삼성전자', 'LG화학', '현대차'}

# 요소 추가
주식_세트.add('카카오')
print(주식_세트)
````

````python
# 요소 제거
주식_세트.remove('LG화학')
print(주식_세트)
````
