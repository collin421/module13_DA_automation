## **1. 리스트란 무엇인가?**

#### **리스트의 정의**
- 리스트는 여러 개의 데이터를 순차적으로 저장할 수 있는 파이썬의 데이터 구조임.
- 리스트는 대괄호 [] 안에 데이터를 **쉼표 ,**로 구분하여 저장함.
- 리스트는 다양한 데이터 타입을 동시에 저장할 수 있음.

#### **리스트 특징**
- 순서가 있음: 리스트의 각 요소는 인덱스를 가지고 있으며, 순서를 유지함.
- 변경 가능: 리스트의 요소는 생성 후에 변경할 수 있음

## **2. 리스트 생성 방법**

```python
# 주식종목 리스트
주식종목_리스트 = ['삼성전자', 'LG화학', '현대차', 'NAVER', '카카오', 'SK하이닉스', 'POSCO']
```

```python
# 금리 정보 리스트 (기준금리, 대출금리, 예금금리)
금리_리스트 = [1.25, 3.6, 2.1]
```

```python
# 혼합된 리스트 (주식종목, 시가총액)
혼합_리스트 = ['삼성전자', 350, '현대차', 50, 'NAVER', 40]
혼합_리스트2 = ['삼성전자', 350.12, '현대차', 50, 'NAVER', 400.0]
```

```python
print(주식종목_리스트)
print(금리_리스트)
print(혼합_리스트)
print(혼합_리스트2)
```

## **3. 리스트 요소 다루기**

**인덱싱과 슬라이싱**
- 인덱싱: 리스트의 각 요소는 인덱스 번호를 가지고 있으며, 이 번호를 통해 특정 요소에 접근할 수 있음. 인덱스는 0부터 시작함.
- 슬라이싱: 리스트의 일부 구간을 잘라내어 새로운 리스트로 반환할 수 있음

```python
주식종목_리스트 = ['삼성전자', 'LG화학', '현대차', 'NAVER', '카카오', 'SK하이닉스', 'POSCO']
```

```python
# 인덱싱
print(주식종목_리스트[1])
print(주식종목_리스트[-1])
```

```python
# 슬라이싱
print(주식종목_리스트[2:5])
print(주식종목_리스트[:3])
print(주식종목_리스트[4:])
```

```python
# index() 데이터의 인덱스 출력
주식종목_리스트.index("NAVER")
```

## **4. 리스트 요소 추가/삭제**

#### **추가**
- append(): 리스트의 끝에 새로운 요소를 추가함.
- insert(): 특정 위치에 새로운 요소를 삽입함.

```python
주식종목_리스트 = ['삼성전자', 'LG화학', '현대차']
```

```python
# 요소 추가 (끝에 추가)
주식종목_리스트.append('카카오')
print(주식종목_리스트)
```

```python
# 특정 위치에 요소 삽입
주식종목_리스트.insert(2, 'NAVER')  # 2번 인덱스에 'NAVER' 삽입
print(주식종목_리스트)
```

#### **삭제**
- del 리스트[인덱스] : 리스트의 특정 인덱스 제거
- remove(): 리스트에서 특정 값을 찾아 제거
- pop(): 리스트의 마지막 요소를 제거하고 반환. 특정 인덱스의 요소를 제거할 수도 있음

```python
주식종목_리스트 = ['삼성전자', 'LG화학', 'NAVER', '현대차', '카카오']
```

```python
# 특정 값 제거
주식종목_리스트.remove('NAVER')
print(주식종목_리스트)
```

```python
# 마지막 요소 제거
주식종목_리스트.pop()
print(주식종목_리스트)
```

```python
# 인덱스를 지정하여 요소 제거
주식종목_리스트.pop(2)  # 2번 인덱스 요소 제거
print(주식종목_리스트)
```

```python
# 인덱스를 지정하여 요소 제거2
del 주식종목_리스트[0]
print(주식종목_리스트)
```

## **5. 리스트의 유용한 함수**

#### **리스트 요소 갯수, 리스트 길이**

```python
# len() 몇개의 리시트가 있는지, 리스트 길이

주식종목_리스트 = ['삼성전자', 'LG화학', '현대차', 'NAVER', '카카오']
print(len(주식종목_리스트))
```

#### **min max 최댓값, 최솟값**
- min() : 최솟값
- max() : 최댓값

```python
주가 = [46000, 57330, 48900, 53200, 50100, 52300]
```

```python
최댓값 = min(주가)
최솟값 = max(주가)
print(최댓값, 최솟값)
```

```python
sum(주가)
```

#### **리스트 정렬**
- sort() 메소드 : 기본 오름차순 정렬, reverse= True : 내림차순 
- sorted() 함수 : reverse = True : 내림차순, False : 오른차순
- reverse() 함수 : 기존 순서 뒤집기

```python
# 리스트 정렬하기, sort()
금리 = [2, 3, 1, 4, 5]
```

```python
# 오름차순 정렬
금리.sort()
print(금리)
```

```python
# 오름차 순
sorted(금리, reverse=False)
```

```python
# 내림차 순
sorted(금리, reverse=True)
```

```python
# 내림차순 정렬
금리.sort(reverse=True)
print(금리)
```

```python
# 순서 뒤집기
금리.reverse()
print(금리)
```

#### **리스트 결합**
- "+" 연산자나 extend()를 사용하여 두 리스트를 결합할 수 있음.

```python
리스트1 = ['삼성전자', 'LG화학']
리스트2 = ['현대차', '카카오']
```

```python
# + 연산자로 리스트 결합
결합_리스트 = 리스트1 + 리스트2
print(결합_리스트)
```

```python
# extend() 메서드로 리스트 결합
리스트1.extend(리스트2)
print(리스트1)
```

#### **문제1) 다음과 같은 휘발유 가격을 리스트로 구성하고, 최댓값과 최솟값을 출력.**

- 2월 4주: 1463원
- 3월 1주: 1483원
- 3월 2주: 1498원
- 3월 3주: 1517원
- 3월 4주: 1530원

```python
휘발유_가격 = [1463, 1483, 1498, 1517, 1530]
최댓값 = max(휘발유_가격)
최솟값 = min(휘발유_가격)
print(f"최댓값: {최댓값}원, 최솟값: {최솟값}원")
```

#### **문제2) 주간 광고 클릭률 데이터를 리스트로 구성하고, 최대 클릭률이 발생한 주차를 출력**

- 1주차: 2.5%
- 2주차: 3.0%
- 3주차: 3.5%
- 4주차: 4.1%
- 5주차: 3.8%

```python
클릭률 = [2.5, 3.0, 3.5, 4.1, 3.8]
최댓값 = max(클릭률)
최댓값_주차 = 클릭률.index(최댓값) + 1
print(f"최대 클릭률이 발생한 주차: {최댓값_주차}주차")
```
