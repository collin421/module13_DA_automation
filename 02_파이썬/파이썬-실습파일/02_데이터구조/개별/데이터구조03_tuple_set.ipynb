## **1. 튜플(Tuple)**
#### **튜플의 정의**
- 순서가 있는 자료형으로, 리스트와 유사하지만 **변경 불가능(Immutable)**함.
- 대괄호가 아닌 **소괄호 ()**를 사용하여 여러 개의 데이터를 저장함.
#### **튜플의 특징**
- 변경 불가: 한 번 정의된 튜플의 요소들은 수정, 삭제, 추가할 수 없음.
- 순서가 있음: 리스트처럼 인덱스를 통해 각 요소에 접근 가능.
#### **튜플의 장점**
- 안정성: 튜플은 수정할 수 없기 때문에, 데이터 보호가 필요한 경우 적합함.
- 메모리 효율: 리스트보다 메모리를 적게 사용함.

```python
# 주식 종목과 시가총액을 튜플로 저장
주식_튜플 = ('삼성전자', 350, '현대차', 50, 'LG화학', 45)

# 튜플 요소 접근
print(주식_튜플[0])
print(주식_튜플[1])
```

```python
# 금융 상품 코드 (변경되지 않는 정보)
금융_상품_코드 = ('KR7005930003', 'KR7005380001', 'KR7000270005')
금융_상품_코드
```

```python
# 주식 정보 튜플
주식_정보 = ('삼성전자', 350)

# 튜플 언패킹
종목명, 시가총액 = 주식_정보
print(종목명)
print(시가총액)
```

```python
# 튜플 생성
주식_튜플 = ('삼성전자', 'LG화학', '현대차')

# 튜플의 요소를 수정하려고 시도
주식_튜플[0] = '카카오'
```

```python
# 튜플 생성
주식_튜플 = ('삼성전자', 'LG화학', '현대차')

# 튜플을 리스트로 변환
주식_리스트 = list(주식_튜플)

# 리스트에서 수정
주식_리스트[0] = '카카오'

# 다시 튜플로 변환
주식_튜플 = tuple(주식_리스트)

print(주식_튜플)
```

## **2. 세트(Set)**
#### **세트의 정의**
- **세트(Set)**는 파이썬의 순서가 없는 자료형으로, 중복을 허용하지 않음.
- **중괄호 {}**를 사용하여 데이터 저장.
#### **세트의 특징**
- 중복 없음: 세트는 중복된 값을 허용하지 않음.
- 순서 없음: 세트는 요소의 순서가 보장되지 않음.
- 수학적 연산: 세트는 교집합, 합집합, 차집합 등 수학적 연산을 쉽게 수행할 수 있음.

```python
# 주식 종목 세트
주식_세트 = {'삼성전자', '현대차', 'LG화학', '삼성전자'}

print(주식_세트)
```

#### **교집합 (&) : 두 세트에 공통으로 포함된 요소를 반환함**

```python
주식_세트1 = {'삼성전자', 'LG화학', '현대차'}
주식_세트2 = {'카카오', '삼성전자', '현대차'}

교집합 = 주식_세트1 & 주식_세트2

print(교집합)
```

#### **합집합 (|) : 두 세트에 포함된 모든 요소를 반환하며, 중복은 제거됨.**

```python
합집합 = 주식_세트1 | 주식_세트2
print(합집합)
```

#### **차집합 (-) : 첫 번째 세트에는 있지만 두 번째 세트에는 없는 요소들을 반환함.**

```python
차집합 = 주식_세트1 - 주식_세트2
print(차집합)
```

#### **세트의 주요 메서드**
- add(): 세트에 새로운 요소를 추가함.
- remove(): 세트에서 특정 요소를 제거함.

```python
주식_세트 = {'삼성전자', 'LG화학', '현대차'}

# 요소 추가
주식_세트.add('카카오')
print(주식_세트)
```

```python
# 요소 제거
주식_세트.remove('LG화학')
print(주식_세트)
```
