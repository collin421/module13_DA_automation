#### **1. 딕셔너리란 무엇인가?**
- 딕셔너리 정의: 키와 값 쌍으로 데이터를 저장하는 파이썬의 데이터 구조.
- 딕셔너리는 대괄호 {} 안에 key: value 형식으로 데이터를 저장하고, 각 키는 고유하며 중복될 수 없음.
- 예시: {"삼성전자": 80000, "현대차": 200000, "카카오": 120000}

#### **2. 딕셔너리 생성 방법**
- 딕셔너리 기본 생성
- dict() 함수나 {}로 빈 딕셔너리를 생성하고 값을 추가

````python
주식_가격 = {}
주식_가격["삼성전자"] = 80000
주식_가격["현대차"] = 200000
주식_가격
````

#### **3. 딕셔너리 요소 다루기**
- 키로 값 접근하기 : 딕셔너리에서 키를 사용하여 특정 값을 가져올 수 있음.

````python
주식_가격 = {"삼성전자": 80000, "현대차": 200000, "카카오": 120000}
주식_가격["삼성전자"]
````

- 키와 값 추가 및 수정 : 딕셔너리에 새로운 키와 값을 추가하거나 기존 키의 값을 수정 가능.

````python
주식_가격["LG화학"] = 150000  # LG화학 추가
주식_가격["삼성전자"] = 82000  # 삼성전자 값 수정
````

#### **4. 딕셔너리 요소 삭제하기**
**요소 삭제 방법**
- del 키워드를 사용하여 특정 키-값 쌍 삭제.
- pop() 메서드를 사용하여 키로 값을 삭제하고 반환

````python
del 주식_가격["현대차"]  # 현대차 제거
주식_가격
````

````python
주식_가격.pop("카카오")  # 카카오 제거 및 값 반환
주식_가격
````

#### **5. 딕셔너리 유용한 메서드**
- keys(), values(), items() : 딕셔너리의 키, 값, 키-값 쌍을 각각 반환.

````python
print(주식_가격.keys())     # 모든 키 반환
print(주식_가격.values())   # 모든 값 반환
print(주식_가격.items())    # (키, 값) 쌍 반환
````

- get() : 키로 값을 가져오며, 키가 없을 경우 None 또는 지정한 기본값을 반환.

### **6. 딕셔너리 활용 실습 문제**

**문제1: 과일_가격 딕셔너리를 만들고, 과일 이름을 키로, 가격을 값으로 저장하시오. 각 과일의 가격을 출력하시오.**
- 예시 데이터: {"사과": 3000, "바나나": 2000, "포도": 4500}

**문제2: 월별 매출 데이터를 딕셔너리로 저장하고, 가장 높은 매출과 낮은 매출을 출력하시오.**
- 예시 데이터: {"1월": 300000, "2월": 280000, "3월": 350000, "4월": 400000}
- list() 함수 - 데이터 구조를 리스트로 바꿔주는 함수

<details>
<summary>✅ 풀이 보기</summary>

```python
# 문제1
과일_가격 = {"사과": 3000, "바나나": 2000, "포도": 4500}
for 과일, 가격 in 과일_가격.items():
    print(f"{과일}: {가격}원")

# 문제2
매출 = {"1월": 300000, "2월": 280000, "3월": 350000, "4월": 400000}
최고 = max(매출, key=매출.get)
최저 = min(매출, key=매출.get)
print(f"가장 높은 매출: {최고} ({매출[최고]:,}원)")
print(f"가장 낮은 매출: {최저} ({매출[최저]:,}원)")
```

**문제 4. 다음의 딕셔너리에서 key 값으로만 구성된 리스트를 생성**

````python
icecream = {'탱크보이': 1200, '폴라포': 1200, '빵빠레': 1800, '월드콘': 1500, '메로나': 1000}
````

- values 값으로만 구성된 리스트를 생성
- icecream 딕셔너리에서 아이스크림 판매 금액의 총합을 출력
- new_product 딕셔너리를 icecream 딕셔너리에 추가
      - new_product = {'팥빙수':2700, '아맛나':1000}

<details>
<summary>✅ 풀이 보기</summary>

```python
icecream = {'탱크보이': 1200, '폴라포': 1200, '빵빠레': 1800, '월드콘': 1500, '메로나': 1000}

# 문제4) key 값으로만 구성된 리스트
키리스트 = list(icecream.keys())
print(키리스트)

# values 값으로만 구성된 리스트
값리스트 = list(icecream.values())
print(값리스트)

# 판매 금액의 총합
print(sum(icecream.values()))

# new_product 추가
new_product = {'팥빙수': 2700, '아맛나': 1000}
icecream.update(new_product)
print(icecream)
```
