#### **문제 1 ) 주어진 주식 가격 리스트의 평균을 계산하고 출력**
- 주식가격 = [1000, 1200, 1300, 1100, 1400]

```python
# 주식 가격 리스트
주식가격 = [1000, 1200, 1300, 1100, 1400]
```

```python
# 주식 가격의 총합을 계산
총합 = 0
for 가격 in 주식가격:
    총합 += 가격  # 각 주식 가격을 총합에 더함

# 평균 계산 (총합 / 요소 개수)
평균가격 = 총합 / len(주식가격)
print(f"평균 주식 가격: {평균가격}원")
```

#### **문제 2 ) 주어진 문자열에서 가장 많이 등장하는 문자를 찾아 출력**
- 문자열 = "hello world, this is a test string"

```python
# 문자열
문자열 = "hello world, this is a test string"
```

```python
# 문자 빈도수 계산
문자빈도 = {}
for 문자 in 문자열:
    if 문자 in 문자빈도:
        문자빈도[문자] += 1
    else:
        문자빈도[문자] = 1

# 빈도 수로 정렬 후 두 번째로 많이 등장하는 문자 찾기
정렬된_문자빈도 = sorted(문자빈도.items(), key=lambda x: (-x[1], x[0]))
두번째_문자, 두번째_횟수 = 정렬된_문자빈도[1]

print(f"두 번째로 많이 등장한 문자: {두번째_문자}, 등장 횟수: {두번째_횟수}")
```

#### **문제3) 주어진 리스트에서 중복된 요소를 모두 제거하고, 중복이 제거된 리스트를 출력**
- 숫자목록 = [1, 2, 2, 3, 4, 4, 5, 6, 6, 7, 8, 8, 9]

```python
# 중복된 요소가 있는 리스트
숫자목록 = [1, 2, 2, 3, 4, 4, 5, 6, 6, 7, 8, 8, 9]
```

```python
# 중복이 제거된 리스트를 저장할 리스트
중복제거목록 = []

# 중복된 요소를 제거
for 숫자 in 숫자목록:
    if 숫자 not in 중복제거목록:
        중복제거목록.append(숫자)

print(f"중복 제거 후: {중복제거목록}")
```
