# **while 문**
- while 문은 조건이 참(True)인 동안 반복적으로 코드를 실행하는 반복문임.
- 조건이 거짓(False)이 되면 반복을 멈춤.
- 반복 횟수가 정해지지 않았을 때 주로 사용함

**언제 사용하는가?**
- 특정 조건을 만족할 때까지 반복을 계속하고 싶을 때.
- 반복 횟수가 미리 정해져 있지 않을 때.
- 주로 이벤트 발생 대기, 무한 루프, 사용자 입력 처리 등에 사용

| 비교 항목    | while 문                        | for 문                                  |
|--------------|----------------------------------|-----------------------------------------|
| **사용 목적**| 조건이 참일 동안 반복           | 반복 가능한 객체 순회 또는 정해진 횟수 반복 |
| **종료 시점**| 조건이 거짓일 때                | 시퀀스나 범위의 끝에 도달할 때          |
| **반복 횟수**| 반복 횟수가 정해져 있지 않을 때 사용 | 반복 횟수가 정해졌을 때 사용             |
| **유형**     | 조건 기반 반복                  | 범위나 컬렉션 기반 반복  

### **for문과 while문으로 1~5까지 출력하기**

```python
# for문
for i in range(1,6):
    print(i)
```

```python
num = 1

while num <= 5:
    print(num)
    num = num + 1 # num += 1
```

### **dict / 첫 번째로 80점 이상인 학생을 찾으면 그 학생의 정보를 출력**
**학생점수 = {"수철": 75, "철수": 70, "미연": 66, "현수": 85, "영희": 90, "민수": 78, "지영": 88}**

```python
학생점수 = {"수철": 75, "철수": 70, "미연": 66, "현수": 85, "영희": 90, "민수": 78, "지영": 88}
```

```python
for 학생, 점수 in 학생점수.items():
    if 점수 >= 80:
        print(f"첫 번째 80점 이상인 학생 : {학생}, 점수 : {점수}")
        break
```

```python
# 학생들 key를 리스트로 만들기
학생들 = list(학생점수.keys())

# 변수 설정
i = 0

# while문 학생과 학생 점수를 추출, 점수가 80점 이상이면 출력하고 반목문 탈출
while i < len(학생들):
    학생 = 학생들[i]
    점수 = 학생점수[학생]
    
    if 점수 >= 80:
        print(f"첫 번째로 80점 이상인 학생: {학생}, 점수: {점수}")
        break
    i += 1
```

#### **문제 1 : while 문을 사용하여 현재 주가가 20,000원이고 5일 연속 상한가(30%)를 기록한 경우의 금액 출력**

```python
주가 = 20000
day = 0

while day < 5:
    주가 = 주가 * 1.3
    print(day,주가)
    day = day + 1
```

#### **문제 2 : 로또 번호 1~45의 숫자 6개를 생성해보세요.**
- 숫자는 중복될 수 없습니다.
- random 모듈, randint(최소 숫자, 최대 숫자 ) = 최소 ~ 최대 숫자 사이에서 랜덤한 숫자 출력

```python
# random 모듈 임포트, randint(최소 숫자, 최대 숫자 ) = 최소 ~ 최대 숫자 사이에서 랜덤한 숫자 출력
import random
number = random.randint(1, 45)
print(number)
```

```python
# for 문일경우 : 반복된 숫자 발생

# 빈 리스트 만들기
lotto = []

# 6번 반복하는 반복문
for i in range(6):
    # 랜덤한 숫자를 변수에 담기
    number = random.randint(1, 45)
    # 로또 리스트에 숫자 추가하기
    lotto.append(number)

print(lotto)
```

```python
# 빈리스트 만들기
lotto = []

# while 반목문 만들기
while len(lotto) < 6:
    number = random.randint(1, 45)
    # 로또 리스트안에 동일한 숫자가 없을 경우 리스트에 추가
    if number not in lotto:
        lotto.append(number)

print(lotto)
```

#### **문제 3 : 1~100까지 랜덤 숫자를 업/다운 게임 만들기**
- 1~100까지 랜덤 숫자 생성
- 사용자로부터 값을 입력 받은 후 사용자가 입력한 숫자가, 랜덥으로 뽑은 값 보다 크면 up, 작으면 down, 맞추면 성공 출력
- input(), input으로 입력 받은 데이터는 문자 타입

```python
# 사용자 입력 받기
# input(), input으로 입력 받은 데이터는 문자 타입

숫자 = input('숫자를 입력하세요 : ')
print(type(숫자), 숫자)
```

```python
# 모듈 임포트
import random

# 랜덤 숫자 생성
num = random.randint(1, 100)
print(num)

while True:
    user_num = input('숫자 입력 :')
    user_num = int(user_num)
    if user_num < num:
        print('down')
    elif user_num == num:
        print('성공')
        break
    else :
        print('up')
```
