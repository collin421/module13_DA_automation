# **1) for_if 중첩**
- 반복되는 데이터 중에서 특정 조건을 만족하는 경우에만 실행하고자 할 때 사용함

### **실습 : 3글자 이상의 과일만 출력하기**
- 과일목록 = ['참외', '사과', '바나나', '산딸기', '블루베리', '복숭아', '키위']

#### **1) 단계 for문을 사용해서 과일 목록 출력하기**

```python
과일목록 = ['참외', '사과', '바나나', '산딸기', '블루베리', '복숭아', '키위']
```

### **실습 : Per이 10미만인 종목 출력하기**
- 종목 = [ ('삼성전자', 15.75), ('LG전자', 308.67),  ('현대차', 8.51), ('NAVER', 55.82), ('SK', 9.15), ('kakao', 8.55)]
- 1. 종목명만 출력하기
  2. per 출력하기
  3. per이 10미만인 종목 출력하기
- PER(Price to Earnings Ratio, 주가수익비율)은 기업의 현재 주가가 주당순이익(EPS)에 비해 몇 배가 되는지를 나타내는 지표
- PER = 현재 주가 / 주당순이익(EPS)
    - 현재 주가: 시장에서 거래되는 기업의 주식 한 주당 가격
    - EPS (Earnings Per Share): 기업의 순이익을 발행된 주식 수로 나눈 값

```python
# 리스트 만들기
data = [
    ('삼성전자', 15.75), 
    ('LG전자', 308.67), 
    ('현대차', 8.51), 
    ('NAVER', 55.82), 
    ('SK', 9.15),
    ('kakao', 8.55)
    ]
```

```python
# 인덱싱해서 출력해보기
print(data[0])
print(data[1])
print(data[2])
print(data[3])
```

```python
# 인덱싱해서 종목명과 per 출력해보기
print(data[0][0])
print(data[0][1])
```

```python
for i in data:
    # print(i[1])
    if i[1] < 10:
        print(i[0])
```

### **실습 : Per이 10미만인 종목만 모으기**

```python
# 빈 리스트 만들기
# 튜플은 수정 불가
result = []
```

```python
for i in data:
    if i[1] < 10: 
        result.append(i[0])

print(result)
```

```python
portfolio = {
    '마소' : [1500, 2300],
    '테슬라' : [1100, 1130],
    '애플' : [600, 530],
    '메타' : [200, 330]
    }
```

```python
portfolio['메타']
```

```python
for k, v in portfolio.items():
    name = k
    매수가 = v[0]
    매도가 = v[1]
    수익율 = (매도가 - 매수가) / 매수가 *100

    if 수익율 >= 5:
        print(name) 
```

# **2) 반복문 탈출 및 뛰어넘기**
#### **break** : 특정 조건을 만족할 때 반복문 탈출하기

**A) 0~9까지를 출력하고 6보다 큰 숫자일 경우 그만 출력하기**

```python
# break를 만나면 반목문을 탈출함
for i in range(10):
    if i > 6:
        break
    print(i)
print("반복문 탈출")
```

**B) 아래 과일목록을 과일바구니에 차례데로 담고, 바나나까지만 담기**
- 과일목록 = ['사과', '체리', '바나나', '파인애플', '자두']

```python
# 아래 과일을 과일바구니에 담고 for문을 탈출
과일목록 = ['사과', '체리', '바나나', '파인애플', '자두']
```

```python
과일바구니 = []
for 과일 in 과일목록:
    과일바구니.append(과일)
    if 과일 == '바나나':
        break
과일바구니
```

## **뛰어넘기** 
#### **continue**
    - 반복문에서 현재 반복을 건너뛰고 다음 반복으로 넘어갈 때 사용함.
    - 특정 조건을 만족할 때 해당 반복만 생략하고, 전체 반복은 중단하지 않음.

#### **A) 0~9에서 홀수만 출력하기**

```python
# 홀수만 출력하기
for i in range(10):
    if i % 2 == 0:
        continue
    print(i)
```

### **문제 1: 상품 재고 관리 시스템**
**A) 상품 리스트에서 재고가 부족한 상품을 찾으면 그 상품의 이름을 출력하고 더 이상 확인하지 않도록 하세요.**
- products = [ {"name": "노트북", "stock": 10}, {"name": "태블릿", "stock": 5}, {"name": "스마트폰", "stock": 0},{"name": "모니터", "stock": 15}]

```python
products = [
    {"name": "노트북", "stock": 10},
    {"name": "태블릿", "stock": 5},
    {"name": "스마트폰", "stock": 0},
    {"name": "모니터", "stock": 15},
]
```

```python
for i in products:
    print(i)
```

```python
for product in products:
    if product["stock"] == 0:
        print(f"{product['name']}의 재고가 부족합니다.")
        break  # 재고 부족한 상품을 찾으면 중단
    print(f"{product['name']}의 재고는 {product['stock']}개 남았습니다.")
```

**B) 과자를 처음부터 순서대로 장바구니에 담고,</br>장바구니에 담긴 과자의 총 가격이 8000원이 초과하면 더 이상 담지 않도록 하기**

- 과자목록 = {'꼬깔콘' : 1500, '포카칩' : 2000, '새우깡' : 1300, '프링클스' : 3000, '스윙칩' : 1500, '꿀꽈배기' : 1200}

```python
과자목록 = {
    '꼬깔콘' : 1500, 
    '포카칩' : 2000, 
    '새우깡' : 1300, 
    '프링클스' : 3000, 
    '스윙칩' : 1500, 
    '꿀꽈배기' : 1200
    }
```

```python
장바구니 = []
총가격 = 0

for 과자, 가격 in 과자목록.items():
    if 총가격 + 가격 >= 8000:
        break
    장바구니.append(과자)
    총가격 = 총가격 + 가격
장바구니
```

### **문제 2: 비정상 거래 걸러내기**
- 거래량 데이터에서 비정상적인 값(예: 음수 또는 결측값)을 제외하고 유효한 거래량만 출력하기

```python
거래량 = [1200, None, 500, -300, 800, 1000, None, 1500]
```

```python
for 거래 in 거래량:
    if 거래 is None or 거래 < 0:  # 결측값이나 음수 값을 건너뜀
        continue
    print(f"유효한 거래량: {거래}주")
```

- 자동차 번호판에서 01~69까지는 승용차를 의미, 다음 자동차 번호판에서 승용차만 출력

```python
번호리스트 = ['10가 1212', 
             '02다 1191', 
             '80가 7788', 
             '91오 7896', 
             '55허 1234', 
             '99하 2323'] 
```

```python
for 번호 in 번호리스트:
    두자기 = 번호[0:2]
    if int(두자기) >= 70:
        continue
    print(번호)
```
