# **조건문 if문과 booltype**

### **1) boolean 타입 데이터**
- 불리언(Boolean) 타입은 참(True) 또는 거짓(False) 값을 가지는 데이터 타입임.
- 파이썬에서 불리언 타입은 주로 조건문이나 비교연산의 결과로 나옴.
- 불리언 값은 논리적 참(True)과 거짓(False)을 표현하며, 프로그램의 흐름을 제어하는 데 사용됨

```python
a = True  # 참
b = False  # 거짓
print(a, type(a))
print(b, type(b))
```

### **2) 비교 연산자**
- 두 값을 비교하여 결과를 True 또는 False로 반환하는데, 이 결과는 불리언 값으로 표현됨.

| 연산자 | 설명         | 예시      | 결과   |
|:-------|:-------------|:----------|:-------|
| ==     | 같다         | 5 == 5    | True   |
| !=     | 같지 않다    | 5 != 3    | True   |
| >      | 크다         | 7 > 3     | True   |
| <      | 작다         | 4 < 6     | True   |
| >=     | 크거나 같다  | 5 >= 5    | True   |
| <=     | 작거나 같다  | 3 <= 4    | True   |

```python
a = 10
b = 5
```

```python
# 값이 같은지 비교 (==)
print(a == b)  # 출력: False (10은 5와 같지 않다)
```

```python
# 값이 다른지 비교 (!=)
print(a != b)  # 출력: True (10은 5와 다르다)
```

```python
# a가 b보다 큰지 비교 (>)
print(a > b)   # 출력: True (10은 5보다 크다)
```

```python
# a가 b보다 작은지 비교 (<)
print(a < b)   # 출력: False (10은 5보다 작지 않다)
```

```python
# a가 b보다 크거나 같은지 비교 (>=)
print(a >= b)  # 출력: True (10은 5보다 크거나 같다)
```

```python
# a가 b보다 작거나 같은지 비교 (<=)
print(a <= b)  # 출력: False (10은 5보다 작지 않다)
```

#### 예시

```python
현재_주식_가격 = 150000
목표_주식_가격 = 160000

# 목표 가격 도달 여부 확인
if 현재_주식_가격 >= 목표_주식_가격:
    print("목표가에 도달했습니다.")
else:
    print("목표가에 도달하지 않았습니다.")
```

## **3)조건문(Conditional Statement)**
- 조건을 확인하고, 조건에 따라 서로 다른 코드를 실행하는 문법임. if, elif, else 키워드를 사용하여 작성하며, 코드의 흐름을 제어

#### **구조**
- if: 조건이 **참(True)** 이면 해당 블록을 실행.
- elif: 첫 번째 조건이 **거짓(False)** 일 때, 다른 조건을 확인하고, 참이면 해당 블록을 실행.
- else: 위의 모든 조건이 **거짓(False)** 이면, 마지막에 실행할 블록.

#### **if_문**

```python
'''if 조건식:
    실행할 코드  # 조건식이 True일 때 실행'''
```

#### **if_만약에 주가가 60,000원 이하이면 매수**

```python
주가 = 61000

print(주가 >= 60000)

if 주가 >= 60000:
    print('매수')
```

#### **if_else**

```python
'''if 조건식:
    실행할 코드  # 조건식이 True일 때 실행
else:
    실행할 코드  # 조건이 False일 때 실행'''
```

#### **if_금액이 무료 배송 50000 기준보다 크거나 같을 때 무료 배송이 적용**

```python
고객1 = 30000
고객2 = 60000

무료배송_기준 = 50000

print(고객1 >= 무료배송_기준)

if 고객1 >= 무료배송_기준:
    print("무료 배송이 적용됩니다.")
else:
    print("무료 배송이 적용되지 않습니다.")
```

#### **if_elif_else**

```python
'''if 조건식:
    실행할 코드  # 조건식이 True일 때 실행
elif 다른조건식:
    실행할 코드  # 첫 번째 조건이 False이고, 이 조건이 True일 때 실행
else:
    실행할 코드  # 모든 조건이 False일 때 실행행'''
```

#### **구매 금액이 100,000원 이상 구매 시 사은품과 무료 배송, 50,000이상이면 무료 배송, 미만일 경우 혜택 없음**
- 구매 금액 : 60000

```python
구매금액 = 60000

if 구매금액 >= 100000:
    print("무료 배송 사은품 제공")
elif 구매금액 >= 50000:
    print("무료 배송 제공")
else :
    print("혜택 없음")
```

## **4) 논리 연산자**
- 복잡한 조건을 만들 때 사용, 1개 이상의 조건으로 조건식을 만들때 사용
- and, or, not 세 가지가

#### **AND 연산자**
    - 양쪽 조건이 모두 **참(True)**일 때만 전체 조건이 참이 됨.

#### **사용자가 회원 등급이 "VIP"이고 구매 금액이 100,000원 이상일 때만 혜택을 제공.**

```python
# 사용자가 회원 등급이 "VIP"이고 구매 금액이 100,000원 이상일 때만 혜택을 제공.

회원등급 = "VIP"
구매금액 = 120000

if 회원등급 == "VIP" and 구매금액 >= 100000:
    print("VIP 회원에게 사은품과 무료 배송 제공")  # 출력: VIP 회원에게 사은품과 무료 배송 제공
else:
    print("혜택 없음")
```

#### **OR 연산자**
    - 어느 한쪽 조건이라도 **참(True)**이면 전체 조건이 참

#### **할인 쿠폰을 가진 사용자이거나, 구매 금액이 50,000원 이상인 경우에 무료 배송을 제공.**

```python
# 할인 쿠폰을 가진 사용자이거나, 구매 금액이 50,000원 이상인 경우에 무료 배송을 제공.

할인쿠폰_유무 = True
구매금액 = 40000

if 할인쿠폰_유무 == True or 구매금액 >= 50000:
    print("무료 배송 제공")  # 출력: 무료 배송 제공
else:
    print("무료 배송 제공 안 됨")
```

#### **NOT 연산자**
    - 설명: 참(True)을 거짓(False)으로, 거짓(False)을 참(True)**으로 바꾸는 연산

#### **사용자가 비회원일 때만 추가 혜택을 제공**

```python
# 사용자가 비회원일 때만 추가 혜택을 제공

회원여부 = True

if not 회원여부:  # 회원여부가 False일 때
    print("비회원에게 추가 혜택 제공")
else:
    print("회원은 추가 혜택 없음")  # 출력: 회원은 추가 혜택 없음
```

#### **복합 논리 연산자 사용**
- 회원 등급이 VIP이거나 플래티넘 등급이고, 구매 금액이 50,000원 이상일 때 혜택을 제공

```python
# 회원 등급이 VIP이거나 플래티넘 등급이고, 구매 금액이 50,000원 이상일 때 혜택을 제공

회원등급 = "플래티넘"
구매금액 = 70000

if (회원등급 == "VIP" or 회원등급 == "플래티넘") and 구매금액 >= 50000:
    print("추가 혜택 제공")  # 출력: 추가 혜택 제공
else:
    print("추가 혜택 없음")
```

- 비회원이면서, 구매 금액이 30,000원 이상일 때만 특정 혜택을 제공

```python
# 비회원이면서, 구매 금액이 30,000원 이상일 때만 특정 혜택을 제공

회원여부 = False
구매금액 = 35000

if not 회원여부 and 구매금액 >= 30000:
    print("비회원에게 5% 추가 할인 제공")  # 출력: 비회원에게 5% 추가 할인 제공
else:
    print("추가 할인 없음")
```

#### **논리 연산자의 우선순위**
- not: 가장 높은 우선순위 > and: 중간 우선순위 > or: 가장 낮은 우선순위
- 괄호를 사용하여 우선 순위를 바꿀 수 있음
(not이 먼저 평가되고, 그 다음에 and, 마지막으로 or가 평가)

## **5) in, not in 연산자**
- in과 not in 연산자는 특정 값이 시퀀스(리스트, 문자열, 튜플 등)에 포함되어 있는지 확인할 때 사용

#### **in 연산자**
- in 연산자는 특정 값이 시퀀스 안에 있는지 확인하는 데 사용됨.
- 조건이 참이면 True, 거짓이면 False 반환함

```python
# 리스트 만들기
리스트 = [1, 2, 3, 4, 5]
문자열 = "hello world"
```

**리스트에서 3이 있는지 확인**

```python
# 리스트에서 3이 있는지 확인
if 3 in 리스트:
    print("리스트에 3이 포함되어 있음")  # 출력됨.
```

**문자열에서 'hello'가 포함되어 있는지 확인**

```python
# 문자열에서 'hello'가 포함되어 있는지 확인
if "hello" in 문자열:
    print("문자열에 'hello'가 포함되어 있음")  # 출력됨.
```

#### **not in 연산자**
- not in 연산자는 특정 값이 시퀀스에 포함되어 있지 않은지 확인하는 데 사용됨.
- 조건이 참이면 True, 거짓이면 False 반환함.

**# 리스트에 6이 없는지 확인**

```python
# 리스트에 6이 없는지 확인
if 6 not in 리스트:
    print("리스트에 6이 포함되어 있지 않음")  # 출력됨.
```

**문자열에 'bye'가 없는지 확인**

```python
# 문자열에 'bye'가 없는지 확인
if "bye" not in 문자열:
    print("문자열에 'bye'가 포함되어 있지 않음")  # 출력됨.
```

#### **문제 4: 로그인 검증 (난이도: 상)**
- 등록된 사용자 정보가 **{"id": "user123", "pw": "password123"}일 때**, 사용자 정보가 **"id": "user123", "pw": "password123"** 과 일치하는지 확인하고,
- 일치하면 "로그인 성공함"을, 일치하지 않으면 "로그인 실패함"을 출력함.

```python
등록된_사용자 = {"id": "user123", "pw": "password123"}
아이디 = "user123"
비밀번호 = "password123"

if 아이디 == 등록된_사용자["id"] and 비밀번호 == 등록된_사용자["pw"]:
    print("로그인 성공함")
else:
    print("로그인 실패함")
```
