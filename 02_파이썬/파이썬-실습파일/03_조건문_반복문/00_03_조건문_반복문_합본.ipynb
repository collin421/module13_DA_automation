# **조건문 if문과 booltype**

### **1) boolean 타입 데이터**
- 불리언(Boolean) 타입은 참(True) 또는 거짓(False) 값을 가지는 데이터 타입임.
- 파이썬에서 불리언 타입은 주로 조건문이나 비교연산의 결과로 나옴.
- 불리언 값은 논리적 참(True)과 거짓(False)을 표현하며, 프로그램의 흐름을 제어하는 데 사용됨

```python
a = True  # 참
b = False  # 거짓
print(a, type(a))
print(b, type(b))
```

### **2) 비교 연산자**
- 두 값을 비교하여 결과를 True 또는 False로 반환하는데, 이 결과는 불리언 값으로 표현됨.

| 연산자 | 설명         | 예시      | 결과   |
|:-------|:-------------|:----------|:-------|
| ==     | 같다         | 5 == 5    | True   |
| !=     | 같지 않다    | 5 != 3    | True   |
| >      | 크다         | 7 > 3     | True   |
| <      | 작다         | 4 < 6     | True   |
| >=     | 크거나 같다  | 5 >= 5    | True   |
| <=     | 작거나 같다  | 3 <= 4    | True   |

```python
a = 10
b = 5
```

```python
# 값이 같은지 비교 (==)
print(a == b)  # 출력: False (10은 5와 같지 않다)
```

```python
# 값이 다른지 비교 (!=)
print(a != b)  # 출력: True (10은 5와 다르다)
```

```python
# a가 b보다 큰지 비교 (>)
print(a > b)   # 출력: True (10은 5보다 크다)
```

```python
# a가 b보다 작은지 비교 (<)
print(a < b)   # 출력: False (10은 5보다 작지 않다)
```

```python
# a가 b보다 크거나 같은지 비교 (>=)
print(a >= b)  # 출력: True (10은 5보다 크거나 같다)
```

```python
# a가 b보다 작거나 같은지 비교 (<=)
print(a <= b)  # 출력: False (10은 5보다 작지 않다)
```

#### 예시

```python
현재_주식_가격 = 150000
목표_주식_가격 = 160000

# 목표 가격 도달 여부 확인
if 현재_주식_가격 >= 목표_주식_가격:
    print("목표가에 도달했습니다.")
else:
    print("목표가에 도달하지 않았습니다.")
```

## **3)조건문(Conditional Statement)**
- 조건을 확인하고, 조건에 따라 서로 다른 코드를 실행하는 문법임. if, elif, else 키워드를 사용하여 작성하며, 코드의 흐름을 제어

#### **구조**
- if: 조건이 **참(True)** 이면 해당 블록을 실행.
- elif: 첫 번째 조건이 **거짓(False)** 일 때, 다른 조건을 확인하고, 참이면 해당 블록을 실행.
- else: 위의 모든 조건이 **거짓(False)** 이면, 마지막에 실행할 블록.

#### **if_문**

```python
'''if 조건식:
    실행할 코드  # 조건식이 True일 때 실행'''
```

#### **if_만약에 주가가 60,000원 이하이면 매수**

```python
주가 = 61000

print(주가 >= 60000)

if 주가 >= 60000:
    print('매수')
```

#### **if_else**

```python
'''if 조건식:
    실행할 코드  # 조건식이 True일 때 실행
else:
    실행할 코드  # 조건이 False일 때 실행'''
```

#### **if_금액이 무료 배송 50000 기준보다 크거나 같을 때 무료 배송이 적용**

```python
고객1 = 30000
고객2 = 60000

무료배송_기준 = 50000

print(고객1 >= 무료배송_기준)

if 고객1 >= 무료배송_기준:
    print("무료 배송이 적용됩니다.")
else:
    print("무료 배송이 적용되지 않습니다.")
```

#### **if_elif_else**

```python
'''if 조건식:
    실행할 코드  # 조건식이 True일 때 실행
elif 다른조건식:
    실행할 코드  # 첫 번째 조건이 False이고, 이 조건이 True일 때 실행
else:
    실행할 코드  # 모든 조건이 False일 때 실행행'''
```

#### **구매 금액이 100,000원 이상 구매 시 사은품과 무료 배송, 50,000이상이면 무료 배송, 미만일 경우 혜택 없음**
- 구매 금액 : 60000

```python
구매금액 = 60000

if 구매금액 >= 100000:
    print("무료 배송 사은품 제공")
elif 구매금액 >= 50000:
    print("무료 배송 제공")
else :
    print("혜택 없음")
```

## **4) 논리 연산자**
- 복잡한 조건을 만들 때 사용, 1개 이상의 조건으로 조건식을 만들때 사용
- and, or, not 세 가지가

#### **AND 연산자**
    - 양쪽 조건이 모두 **참(True)**일 때만 전체 조건이 참이 됨.

#### **사용자가 회원 등급이 "VIP"이고 구매 금액이 100,000원 이상일 때만 혜택을 제공.**

```python
# 사용자가 회원 등급이 "VIP"이고 구매 금액이 100,000원 이상일 때만 혜택을 제공.

회원등급 = "VIP"
구매금액 = 120000

if 회원등급 == "VIP" and 구매금액 >= 100000:
    print("VIP 회원에게 사은품과 무료 배송 제공")  # 출력: VIP 회원에게 사은품과 무료 배송 제공
else:
    print("혜택 없음")
```

#### **OR 연산자**
    - 어느 한쪽 조건이라도 **참(True)**이면 전체 조건이 참

#### **할인 쿠폰을 가진 사용자이거나, 구매 금액이 50,000원 이상인 경우에 무료 배송을 제공.**

```python
# 할인 쿠폰을 가진 사용자이거나, 구매 금액이 50,000원 이상인 경우에 무료 배송을 제공.

할인쿠폰_유무 = True
구매금액 = 40000

if 할인쿠폰_유무 == True or 구매금액 >= 50000:
    print("무료 배송 제공")  # 출력: 무료 배송 제공
else:
    print("무료 배송 제공 안 됨")
```

#### **NOT 연산자**
    - 설명: 참(True)을 거짓(False)으로, 거짓(False)을 참(True)**으로 바꾸는 연산

#### **사용자가 비회원일 때만 추가 혜택을 제공**

```python
# 사용자가 비회원일 때만 추가 혜택을 제공

회원여부 = True

if not 회원여부:  # 회원여부가 False일 때
    print("비회원에게 추가 혜택 제공")
else:
    print("회원은 추가 혜택 없음")  # 출력: 회원은 추가 혜택 없음
```

#### **복합 논리 연산자 사용**
- 회원 등급이 VIP이거나 플래티넘 등급이고, 구매 금액이 50,000원 이상일 때 혜택을 제공

```python
# 회원 등급이 VIP이거나 플래티넘 등급이고, 구매 금액이 50,000원 이상일 때 혜택을 제공

회원등급 = "플래티넘"
구매금액 = 70000

if (회원등급 == "VIP" or 회원등급 == "플래티넘") and 구매금액 >= 50000:
    print("추가 혜택 제공")  # 출력: 추가 혜택 제공
else:
    print("추가 혜택 없음")
```

- 비회원이면서, 구매 금액이 30,000원 이상일 때만 특정 혜택을 제공

```python
# 비회원이면서, 구매 금액이 30,000원 이상일 때만 특정 혜택을 제공

회원여부 = False
구매금액 = 35000

if not 회원여부 and 구매금액 >= 30000:
    print("비회원에게 5% 추가 할인 제공")  # 출력: 비회원에게 5% 추가 할인 제공
else:
    print("추가 할인 없음")
```

#### **논리 연산자의 우선순위**
- not: 가장 높은 우선순위 > and: 중간 우선순위 > or: 가장 낮은 우선순위
- 괄호를 사용하여 우선 순위를 바꿀 수 있음
(not이 먼저 평가되고, 그 다음에 and, 마지막으로 or가 평가)

## **5) in, not in 연산자**
- in과 not in 연산자는 특정 값이 시퀀스(리스트, 문자열, 튜플 등)에 포함되어 있는지 확인할 때 사용

#### **in 연산자**
- in 연산자는 특정 값이 시퀀스 안에 있는지 확인하는 데 사용됨.
- 조건이 참이면 True, 거짓이면 False 반환함

```python
# 리스트 만들기
리스트 = [1, 2, 3, 4, 5]
문자열 = "hello world"
```

**리스트에서 3이 있는지 확인**

```python
# 리스트에서 3이 있는지 확인
if 3 in 리스트:
    print("리스트에 3이 포함되어 있음")  # 출력됨.
```

**문자열에서 'hello'가 포함되어 있는지 확인**

```python
# 문자열에서 'hello'가 포함되어 있는지 확인
if "hello" in 문자열:
    print("문자열에 'hello'가 포함되어 있음")  # 출력됨.
```

#### **not in 연산자**
- not in 연산자는 특정 값이 시퀀스에 포함되어 있지 않은지 확인하는 데 사용됨.
- 조건이 참이면 True, 거짓이면 False 반환함.

**# 리스트에 6이 없는지 확인**

```python
# 리스트에 6이 없는지 확인
if 6 not in 리스트:
    print("리스트에 6이 포함되어 있지 않음")  # 출력됨.
```

**문자열에 'bye'가 없는지 확인**

```python
# 문자열에 'bye'가 없는지 확인
if "bye" not in 문자열:
    print("문자열에 'bye'가 포함되어 있지 않음")  # 출력됨.
```

# **for 반복문**
- 반복 가능한 객체(리스트, 문자열, 튜플, 딕셔너리 등)를 순차적으로 순회하며 작업을 반복 수행.
- 정해진 횟수나 컬렉션의 요소를 처리할 때 사용.

```python
# 기본 구조
'''for 변수 in 반복가능한_객체:
    실행할 코드'''
```

- 변수: 반복 가능한 객체의 각 요소를 받아오는 변수임.
- 반복가능한_객체: 리스트, 문자열, 튜플, 딕셔너리 등 순회할 수 있는 객체임.

#### **과일목록 모두 출력하기 사과, 바나나, 체리, 포도, 키위**

```python
# 리스트 타입

과일목록 = ["사과", "바나나", "체리", "포도", "키위"]

for 과일 in 과일목록:
    print(과일)
```

```python
# 튜플 타입
과일목록 = ("사과", "바나나", "체리", "포도", "키위")

for 과일 in 과일목록:
    print(과일)
```

```python
# 문자열
문자열 = "Python"

for 문자 in 문자열:
    print(문자)
```

#### **딕셔너리 타입**
- key로 vlue 인덱싱
- items() 두 개의 변수로 key, value 언패킹

```python
# 딕셔너리 타입
학생점수 = {"철수": 85, "영희": 90, "민수": 78}
```

```python
# key값 출력
for 학생 in 학생점수.keys():
    print(f"{학생}")
```

```python
# key값을 활용해서 value값 출력
for 학생 in 학생점수.keys():
    print(f"{학생점수[학생]}")
```

```python
# key와 value 모두 출력
for 학생, 점수 in 학생점수.items():
    print(f"{학생}의 점수는 {점수}점임.")
```

#### **range(), for문으로 정수 범위를 표현**
- 숫자 범위를 생성하는 함수로, 주로 반복문에서 순차적인 숫자 시퀀스를 생성할 때 사용됨. range()는 시작값, 종료값, 간격을 지정하여 범위를 생성
- 예) 0-100까지 출력, 2002-2040년 까지 4년 단위로 출력

#### 활용
- range(start, end, step)
- range(0,5) : 0에서 4까지 출력
- range(1, 11, 2) : 1에서 10까지 2씩 증가하는 숫자 출력

### **dict / 첫 번째로 80점 이상인 학생을 찾으면 그 학생의 정보를 출력**
**학생점수 = {"수철": 75, "철수": 70, "미연": 66, "현수": 85, "영희": 90, "민수": 78, "지영": 88}**

```python
학생점수 = {"수철": 75, "철수": 70, "미연": 66, "현수": 85, "영희": 90, "민수": 78, "지영": 88}
```

```python
for 학생, 점수 in 학생점수.items():
    if 점수 >= 80:
        print(f"첫 번째 80점 이상인 학생 : {학생}, 점수 : {점수}")
        break
```

```python
# 학생들 key를 리스트로 만들기
학생들 = list(학생점수.keys())

# 변수 설정
i = 0

# while문 학생과 학생 점수를 추출, 점수가 80점 이상이면 출력하고 반목문 탈출
while i < len(학생들):
    학생 = 학생들[i]
    점수 = 학생점수[학생]
    
    if 점수 >= 80:
        print(f"첫 번째로 80점 이상인 학생: {학생}, 점수: {점수}")
        break
    i += 1
```

- 숫자는 중복될 수 없습니다.
- random 모듈, randint(최소 숫자, 최대 숫자 ) = 최소 ~ 최대 숫자 사이에서 랜덤한 숫자 출력

```python
# random 모듈 임포트, randint(최소 숫자, 최대 숫자 ) = 최소 ~ 최대 숫자 사이에서 랜덤한 숫자 출력
import random
number = random.randint(1, 45)
print(number)
```

```python
# for 문일경우 : 반복된 숫자 발생

# 빈 리스트 만들기
lotto = []

# 6번 반복하는 반복문
for i in range(6):
    # 랜덤한 숫자를 변수에 담기
    number = random.randint(1, 45)
    # 로또 리스트에 숫자 추가하기
    lotto.append(number)

print(lotto)
```

```python
# 빈리스트 만들기
lotto = []

# while 반목문 만들기
while len(lotto) < 6:
    number = random.randint(1, 45)
    # 로또 리스트안에 동일한 숫자가 없을 경우 리스트에 추가
    if number not in lotto:
        lotto.append(number)

print(lotto)
```

#### **문제 1 ) 주어진 주식 가격 리스트의 평균을 계산하고 출력**
- 주식가격 = [1000, 1200, 1300, 1100, 1400]

```python
# 주식 가격 리스트
주식가격 = [1000, 1200, 1300, 1100, 1400]
```

```python
# 주식 가격의 총합을 계산
총합 = 0
for 가격 in 주식가격:
    총합 += 가격  # 각 주식 가격을 총합에 더함

# 평균 계산 (총합 / 요소 개수)
평균가격 = 총합 / len(주식가격)
print(f"평균 주식 가격: {평균가격}원")
```

#### **문제 2 ) 주어진 문자열에서 가장 많이 등장하는 문자를 찾아 출력**
- 문자열 = "hello world, this is a test string"

```python
# 문자열
문자열 = "hello world, this is a test string"
```

```python
# 문자 빈도수 계산
문자빈도 = {}
for 문자 in 문자열:
    if 문자 in 문자빈도:
        문자빈도[문자] += 1
    else:
        문자빈도[문자] = 1

# 빈도 수로 정렬 후 두 번째로 많이 등장하는 문자 찾기
정렬된_문자빈도 = sorted(문자빈도.items(), key=lambda x: (-x[1], x[0]))
두번째_문자, 두번째_횟수 = 정렬된_문자빈도[1]

print(f"두 번째로 많이 등장한 문자: {두번째_문자}, 등장 횟수: {두번째_횟수}")
```

#### **문제3) 주어진 리스트에서 중복된 요소를 모두 제거하고, 중복이 제거된 리스트를 출력**
- 숫자목록 = [1, 2, 2, 3, 4, 4, 5, 6, 6, 7, 8, 8, 9]

```python
# 중복된 요소가 있는 리스트
숫자목록 = [1, 2, 2, 3, 4, 4, 5, 6, 6, 7, 8, 8, 9]
```

```python
# 중복이 제거된 리스트를 저장할 리스트
중복제거목록 = []

# 중복된 요소를 제거
for 숫자 in 숫자목록:
    if 숫자 not in 중복제거목록:
        중복제거목록.append(숫자)

print(f"중복 제거 후: {중복제거목록}")
```
