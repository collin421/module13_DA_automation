5. 지역을 입력하면 지역의 vip 고객명단을 추출해서 csv 파일로 만들어주는 코드 작성
6. 연령 범위를 입력하면 해당 연령 구간의 회원을 출력해주고 파일로 만들어주는 코드

```python
# 4. VIP회원 중 총 구매가격이 낮은 상위 10명의 고객 확인하고 csv 파일로 만들기
cond = df['VIP회원'] == 'VIP'
df_sort = df[cond].sort_values(by='총구매금액')
df_sort.head(10)
```

```python
# 5. 지역을 입력하면 지역의 vip 고객명단을 추출해서 csv 파일로 만들어주는 코드 작성
지역 = input('지역을 입력하세요 : ')

cond1 = df['VIP회원'] == 'VIP'
cond2 = df['Address'] == 지역

df.loc[cond1 & cond2].to_csv(f'{지역}_vip.csv')
```

```python
# 6. 연령 범위를 입력하면 해당 연령 구간의 회원을 출력해주고 파일로 만들어주는 코드
```

```python
연령_최소값 = int(input(' 연령 범위를 입력(최소값) : '))
연령_최대값 = int(input(' 연령 범위를 입력(최대값) : '))

cond = df['Age'].between(연령_최소값, 연령_최대값)
df3 = df.loc[cond]
df3.to_csv("연령별_member.csv")
```
