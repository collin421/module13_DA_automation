## **1) 데이터 정렬하기**

#### **한가지 열을을 기준으로 정렬하기**

## **2) 데이터 필터링하기**

### **1. 기본 조건 필터링**

- 단일 조건을 사용하여 특정 열의 값이 주어진 조건을 만족하는 행만 선택할 수 있음
- 불린 데이터와 필터링
    - 불린 데이터는 참(True) 또는 거짓(False) 두 가지 값만을 가짐. 데이터 필터링에서는 각 조건이 불린 값으로 평가되며, True인 경우에만 데이터를 선택함.
- == 같다, != 다르다, >	초과, <	미만, >= 이상, <= 이하

#### **한가지 조건으로 필터하기**

```python
# 조건 연산자로 불린데이터 만들기 
# 현재 주가가 20만원 이상인 데이터의 불린 정보
df['현재가'] >= 200000
```

```python
# 불린 데이터로 행 필터링, 현재 주가가 20만원 이상인 데이터만 필터링
# True 데이터만 필터링 됨
df_filter1 = df[df['현재가']>=200000]
df_filter1
```

```python
# 지역이 서울인 데이터 필터링
df['지역'] == '서울'
```

```python
df_seoul = df[df['지역'] == '서울']
df_seoul
```

```python
# 지역이 서울이 아닌 데이터 필터링
df['지역'] != '서울'
```

```python
df_filter2 = df['지역'] != '서울'
df_filter2
```

#### **여로 조건을 사용한 필터링(AND, OR 조건)**
- 여러 조건을 결합하여 데이터의 필터링을 세분화할 수 있음.
- AND 조건은 & 연산자를 사용하고, OR 조건은 | 연산자를 사용
- 각 조건을 () 괄호로 묶어줌

#### **AND, & : 모두 만족하는 데이터**

```python
# 현재가가 100,000 이상이고 거래량이 500,000 이상인 데이터

# 현재가가 100,000 이상
df[df['현재가'] >= 100000]
```

```python
# 거래량이 500,000 이상
df[df['거래량'] >= 500000]
```

```python
# 두가지 조건을 모두 만족 : AND 조건은 & 연산자를 사용
df_filter3 = df[(df['현재가'] >= 100000) & (df['거래량'] >= 500000)]
df_filter3
```

```python
# 지역이 서울, 거래량이 500,000 이상, 현재가가 50,000원 이상인 데이터
df_filter4 = df[(df['지역'] == '서울') & (df['거래량'] >= 500000) & (df['현재가'] >= 50000)]
df_filter4
```

#### **OR, | : 하나라도 만족하는 데이터**

```python
# '현재가'가 100,000 이상이거나 '거래량'이 500,000 이상인 종목 필터링 (OR 조건)
df_filter5 = df[(df['현재가'] >= 100000) | (df['거래량'] >= 500000)]
df_filter5
```

```python
# 거래량이 500000이상 이거나 지역이 성남인 데이터
df_filter6 = df[(df['지역'] == '성남') | (df['거래량'] >= 500000)]
df_filter6
```

#### **AND OR 혼합해서 쓰기**

```python
# 현재가 100000원 이상 거래량이 500000이상, 또는 지역이 성남인 데이터
df_filter7 = df[(df['지역'] == '성남') | ((df['현재가'] >= 100000) & (df['거래량'] >= 500000))]
df_filter7
```

#### **조건 부정, ~ : (NOT 조건)**

```python
# 서울 지역이인 데이터
df[df['지역']=='서울']
```

```python
# 서울 지역이 아닌 데이터
df[df['지역']!='서울']
```

```python
# 서울 지역이 아닌 데이터
# ~를 통해서 True와 False를 바꿈
df[~(df['지역']=='서울')]
```

```python
df_filter7 = df[~(df['지역']=='서울')]
df_filter7
```

### **2. loc[]** : loc[조건, 열 이름]
- 특정 조건을 만족하는 행만을 선택하고, 필요한 열을 함께 선택할 수 있음

```python
df.columns
```

```python
# 현재가가 200,000 이상인 종목의 종목명, 현재가 열만 필터링
df.loc[df['현재가'] >= 200000, ['종목명', '현재가']]
```

```python
# 복수조건 변수로 설정하기
# '현재가'가 200,000 이상이고 '거래량'이 500,000 이상인 데이터에서 '종목명', '현재가', '거래량' 열 선택
cond1 = (df['현재가'] >= 200000)
cond2 = (df['거래량'] >= 500000)
```

```python
df_filter = df.loc[cond1 & cond2, ['종목명', '현재가', '거래량']]
df_filter
```

```python
#특정열을 선택 후 필터링하기
df[['종목명', '현재가', '거래량']].loc[cond1 & cond2]
```

```python
# 행 필터링 후 특정 열 선택하기
df.loc[cond1 & cond2][['종목명', '현재가', '거래량']]
```

**연습 문제 3**
- 1) 전일대비변동률이 0.5% 이상이고, 거래량이 400,000 이상인 종목을 필터링한 후, 거래량을 기준으로 내림차순 정렬하시오.
- 2) 시가총액이 500,000억원 이상인 종목 중, **'현재가'**가 200,000 이상인 종목을 필터링하고, 현재가 기준으로 내림차순 정렬하시오.
- 3) **업종이 '인터넷'**인 종목 중에서 현재가가 200,000 이상인 종목을 필터링하고, 전일대비변동률(%) 기준으로 내림차순 정렬하시오.

```python
# 1. 전일대비변동률이 0.5% 이상이고, 거래량이 400,000 이상인 종목을 필터링한 후, 거래량을 기준으로 내림차순 정렬
df_filtered_sorted = df[(df['전일대비변동률(%)'] >= 0.5) & (df['거래량'] >= 400000)]
df_filtered_sorted = df_filtered_sorted.sort_values(by='거래량', ascending=False)
df_filtered_sorted
```

```python
# 2. 시가총액이 500,000억원 이상인 종목 중, **'현재가'**가 200,000 이상인 종목을 필터링하고, 현재가 기준으로 내림차순 정렬
df_filtered_sorted = df[(df['시가총액(억원)'] >= 500000) & (df['현재가'] >= 200000)]
df_filtered_sorted = df_filtered_sorted.sort_values(by='현재가', ascending=False)
df_filtered_sorted
```

```python
# 3. 업종이 '인터넷'**인 종목 중에서 현재가가 200,000 이상인 종목을 필터링하고, 전일대비변동률(%) 기준으로 내림차순 정렬
df_filtered_sorted = df[(df['업종'] == '인터넷') & (df['현재가'] >= 200000)]
df_filtered_sorted = df_filtered_sorted.sort_values(by='전일대비변동률(%)', ascending=False)
df_filtered_sorted
```
