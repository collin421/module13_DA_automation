## **Pandas 데이터 요약하기**
- 데이터 요약은 데이터를 이해하고 분석하는 첫 번째 단계로 매우 중요함.
- 데이터셋이 커질수록 개별 데이터를 일일이 확인하기 어렵기 때문에, 요약을 통해 주요 패턴과 특성을 빠르게 파악할 수 있음.
- 이를 통해 데이터의 분포, 통계적 경향 등을 파악하고, 추가 분석 방향을 설정할 수 있음.

### **왜 데이터 요약**
- **1. 데이터의 전반적인 분포 파악:** 수백, 수천 개의 행이 있는 데이터셋에서 중요한 패턴을 빠르게 찾아내기 위해 요약이 필요함.
- **2. 특정 그룹별 통계 분석:** 데이터의 그룹별로 중요한 통계를 계산할 수 있음.
- **3. 데이터 이상치 및 누락 데이터 확인:** 요약을 통해 데이터의 결측값이나 이상치 여부를 빠르게 확인할 수 있음.
- **4. 데이터의 주요 지표 추출:** 평균, 중간값, 최소값, 최대값 등의 주요 지표를 통해 데이터의 핵심 정보를 파악할 수 있음.

### **1) pivot_table() :** 특정 열을 기준으로 데이터를 요약하고 집계할 때 사용됨. 
- index와 values를 지정하여 데이터를 그룹화하고, 각 그룹의 값을 요약
- 피벗 테이블은 엑셀에서 많이 사용되는 개념으로, 판다스에서도 데이터 요약에 매우 유용함.

**주요 인수**
- data: 피벗 테이블을 만들 데이터프레임.
- values: 집계하려는 열. (예: 매출액, 거래량)
- index: 그룹화할 행 기준 열. (예: 제품명, 날짜)
- columns: 그룹화할 열 기준. (예: 분기, 카테고리)
- aggfunc: 적용할 집계 함수. 기본값은 'mean'이며, 'sum', 'count', 'max', 'min' 등의 함수 사용 가능.
- margins: 행과 열의 총계를 포함할지 여부. 기본값은 False.
- margins_name: 총계에 붙일 이름. 기본값은 'All'.

```python
import pandas as pd
```

```python
df = pd.read_csv('./08_pivot_제품.csv', index_col=0)
df
```

**데이터 설명**
- 지역: 제품이 판매된 지역 (서울, 부산).
- 제품: 판매된 제품 이름 (삼성 갤럭시 S21, LG OLED TV, 애플 아이폰 12).
- 분기: 제품이 판매된 분기 (Q1, Q2).
- 판매량: 각 제품의 판매 수량.
- 매출액: 각 제품의 매출 금액 (원).
- 할인율: 판매 시 적용된 할인율 (%).

#### **1. 기본 구조: index와 values 사용**
- pivot_table은 평균을 계산

```python
# 제품별 매출액 평균, 기본 피벗테이블 집계는 평균
pd.pivot_table(data=df, index='제품', values='판매량')
```

#### **Step 2: 집계 방식 변경 (aggfunc)**
- aggfunc 인수를 사용하면 다양한 방식으로 집계

#### **aggfunc 파라미터**

- mean: 평균
- sum: 합계
- min: 최소값
- max: 최대값
- count: 값의 개수
- std: 표준편차
- var: 분산
- median: 중앙값
- first: 첫 번째 값
- last: 마지막 값
- prod: 곱셈
- nunique: 고유값의 개수

**판매량의 합계 계산**

```python
# aggfunc 파라미터를 활용해서 합계로 집계하기
pd.pivot_table(data=df, index='제품', values='판매량', aggfunc='sum')
```

#### **Step 3: 다중 집계 (여러 값과 집계 방식 사용)**
- 여러 개의 values와 집계 함수를 동시에 사용

**판매량과 매출액의 합계 및 평균 계산**

```python
# 제품별 지역별 매출액, 판매량 합계
pd.pivot_table(data=df, index='제품', values=['판매량', '매출액'], aggfunc={'판매량': 'sum', '매출액': 'mean'})
```

#### **Step 4: 다중 인덱스 사용 (index에 여러 열 지정)**
-  index에 여러 열을 지정하면, 데이터를 더 세분화하여 분석

**제품과 지역을 기준으로 판매량과 매출액 분석**

```python
# 데이터 프레임 메소드로 생성하기
pd.pivot_table(data=df, index=['제품', '지역'], values=['판매량', '매출액'], aggfunc='sum')
```

#### **Step 5: 열을 기준으로 데이터 정리 (columns 사용)**
- columns 파라미터를 사용하면 데이터의 특정 열을 **컬럼(열)**으로 변환

**분기를 기준으로 제품 판매량 분석**

```python
pd.pivot_table(data=df, index='제품', values='판매량', columns='분기', aggfunc='sum')
```

#### **Step 6: 총합 추가 (margins 사용)**
- 총합을 계산하여 피벗 테이블에 포함

```python
# margin=True : 총합 , margin_name='총합' 총합 이름 설정
pd.pivot_table(data=df, index='제품', values='판매량', columns='분기', aggfunc='sum', margins=True, margins_name='총합' )
```

```python
# 총합 행 총합 삭제
pivot = pd.pivot_table(data=df, index='제품', values='판매량', columns='분기', aggfunc='sum', margins=True, margins_name='총합' ).drop('총합', axis=0)
pivot
```

#### **데이터 설명**

**렌탈 고객 데이터 정보**
- Index: 각 행의 고유 식별자 (인덱스).
- Member_ID: 고객의 고유 식별자 (회원 번호).
- Sales_Type: 판매 유형. 예: 렌탈, 멤버십 등.
- Contract_Type: 계약 유형. 예: 일반계약, 교체계약, 재계약 등.
- Channel: 판매 채널. 예: 영업방판, 홈쇼핑/방송 등.
- Datetime: 계약 날짜. 계약이 체결된 날짜를 나타냄.
- Term: 계약 기간 (개월 단위). 예: 60개월, 12개월 등.
- Payment_Type: 결제 방식. 예: CMS(자동이체), 카드이체 등.
- Product_Type: 제품 유형. 예: DES-1 (제품 코드).
- Amount_Month: 월별 결제 금액.
- Customer_Type: 고객 유형. 예: 개인.
- Age: 고객 나이.
- Address1: 고객의 주소 (1차 주소).
- Address2: 고객의 주소 (2차 주소).
- State: 계약 상태. 예: 계약확정, 기간만료 등.
- Overdue_count: 연체 횟수.
- Overdue_Type: 연체 유형. 예: 있음, 없음.
- Gender: 고객 성별. 예: 남자, 여자.
- Credit_Rank: 신용 등급.
- Bank: 결제 은행 또는 카드사. 예: 새마을금고, 현대카드 등.

#### **1. 고객들의 유입 채널 별 월랜탈 매출(평균, 합계)**

```python
# 평균
df1.pivot_table(index='Channel', values='Amount_Month')
```

```python
# 합계
df1.pivot_table(index='Channel', values='Amount_Month', aggfunc='sum')
```

#### **2. 고객들의 성별 및 유입 채널 별 월랜탈 매출 및 연체건 수(합계)**

```python
df1.pivot_table(index=['Gender', 'Channel'], values=['Amount_Month', 'Overdue_count'], aggfunc='sum')
```

#### **3. 판매유형별 연간/월별 월랜탈 매출 합계, 엑셀파일로 저장하기**

```python
# 날짜 타입으로 데이터 변환
df1['Datetime(변환)'] = pd.to_datetime(df1['Datetime'])
df1
```

```python
# 연도 및 월 열 추가
df1['연도'] = df1['Datetime(변환)'].dt.year
df1['월'] = df1['Datetime(변환)'].dt.month
df1
```

```python
df1.pivot_table(index='Sales_Type', columns=['연도', '월'],\
                values='Amount_Month', aggfunc='sum').to_excel('연도별월별_월별렌탈비용.xlsx')
```

- b. 렌탈 상품 유형별, 월 렌탈 매출 합계, 총합, 연간으로 나누어서 보기
- b-1. 2019년대비 20년 증감액과 증감율 열 추가

**a. 결재 은행별, 월 렌탈 비용 합계, 총합 테이블**

```python
df1.pivot_table(index='Bank', values='Amount_Month', aggfunc='sum', margins=True)
```

**b. 렌탈 상품 유형별, 월 렌탈 매출 합계, 총합, 연간으로 나누어서 보기**
