## **데이터 구조 파악하기**
**데이터 분석의 출발점으로, 구조 파악 없이는 분석 전 과정에서 오류와 비효율이 발생할 수 있음.**

**1) 적합한 분석 방법 선택**
- 데이터 형식에 따라 적절한 분석 방법이 달라짐.
- 예를 들어, 숫자형 데이터는 평균과 표준편차를, 범주형 데이터는 빈도수를 분석하는 것이 효과적임.

**2) 오류 방지**
- 데이터 구조를 제대로 이해하지 않으면 타입 오류나 결측치로 인한 문제가 발생할 수 있음.
- 예를 들어, 범주형 데이터를 숫자형으로 잘못 처리하면 분석 오류로 이어질 수 있음.

**3) 데이터 전처리 효율성 향상**
- 데이터 구조를 파악하면 결측치, 이상치 등을 미리 처리하여 전처리 시간을 절약하고, 분석 결과의 신뢰도를 높일 수 있음.

## **숫자 데이터 - 기술 통계 분석(DDA)**

- 숫자 데이터는 데이터의 중심 경향, 분포, 변동성 등을 파악하는 것이 중요함. 
- 대표값과 산포도를 확인하여 데이터의 특징을 이해할 수 있음.

**주요 사항**
- 대표값: 평균, 중앙값, 최빈값 등 데이터의 중심을 나타내는 값.
- 산포도: 데이터가 얼마나 흩어져 있는지를 나타내는 값. 분산, 표준편차 등을 확인.
- 분포: 데이터가 어떤 분포를 따르는지 파악. 최소값, 최대값, 분위수 등을 통해 데이터의 범위를 확인.
- 상관관계: 숫자형 데이터 간의 관계를 파악. 상관계수로 두 변수 간의 연관성을 확인.

#### **대표값** : 데이터의 중심을 나타내는 값으로, 데이터의 전반적인 특징을 요약하는 데 사용됨
- **평균 (mean()) :** 데이터의 산술 평균을 계산함. 데이터 전체를 합한 후 데이터 개수로 나눈 값
- **중앙값 (median()) :** 데이터를 크기 순서대로 정렬했을 때 가운데 위치한 값임. 평균이 이상치에 영향을 받을 때, 중앙값은 데이터의 중심 경향을 더 잘 나타냄
- **최빈값 (mode()) :** 데이터에서 가장 자주 등장하는 값임. 특히 범주형 데이터에서 유용하지만, 숫자형 데이터에서도 최빈값을 확인할 수 있음
#### **산포도** : 데이터가 평균을 중심으로 얼마나 흩어져 있는지, 즉 데이터의 변동성을 나타냄. 
- **분산 (Variance, var()) :** 분산은 데이터가 평균으로부터 얼마나 퍼져 있는지를 나타내는 값임. 데이터와 평균의 차이를 제곱하여 평균낸 값임.
- **표준편차 (Standard Deviation, std()) :** 분산의 제곱근을 취한 값으로, 데이터의 흩어짐 정도를 나타냄. 값이 클수록 데이터가 평균에서 많이 퍼져 있음을 의미함
#### **데이터의 범위와 분포** : 숫자 데이터의 최소값, 최대값, 분위수(quantile) 등을 확인하여 데이터가 어떻게 분포되어 있는지 파악함. 
- **최소값 (min, min())**: 데이터에서 가장 작은 값을 반환.
- **최대값 (max, max())**: 데이터에서 가장 큰 값을 반환.
- **분위수 (Quantiles, quantile())** : 데이터의 특정 백분위수를 계산함. 예를 들어, **0.25 분위수(1사분위)**는 데이터의 하위 25%에 해당하는 값임
- **사분위 범위 (IQR - Interquartile Range)** : 사분위 범위는 데이터의 중앙 50%에 해당하는 값의 범위를 나타냄. 3사분위(75%)에서 1사분위(25%)를 뺀 값임
- **첨도(Kurtosis, kurt())** : 데이터 분포의 꼬리 두께를 측정하는 값임. 첨도는 분포가 얼마나 뾰족한지 또는 평평한지를 나타냄. 정규 분포의 첨도는 3에 가까움
- **왜도(Skewness, skew())** : 는 데이터의 비대칭성을 나타내는 통계 값임. 정규 분포는 왜도가 0

### **데이터 범위와 분포**

```python
max = df['Amount_Month'].max()
max
```

```python
# 데이터 범위
range = max - min
range
```

## **범주형 데이터**

- **항목과 빈도**
- 문자 데이터를 확인할 때, 숫자형 데이터와는 다르게 **고유값, 빈도수, 길이**와 같은 특성을 주로 살펴보게 됨.
- 문자 데이터의 데이터 구조를 파악하는 과정에서는 **데이터 타입 확인, 고유값 확인, 빈도수 계산, 문자열 길이** 등을 확인

#### **갯수, 최빈값, 고유값(항목)과 빈도(갯수) 확인**
- count() : 데이터의 갯수
- mode() : 최빈값
- unique() : 고유값 / nunique() : 고유값의 갯수
- value_counts() : 빈도 수(객수) - Frequency / 상대 빈도 : 각 범주의 비율  normalize=True

```python
# count() : 데이터의 개수
df['Bank'].count()
```

```python
# mode()최빈값
df['Bank'].mode()
```

```python
# 고유값 확인 unique(): 특정 열의 고유값
df['Sales_Type'].unique()
```

```python
# 고객들이 어떤 수단으로 결재하는지, 어떤 은행으로 결재하는 지
df['Bank'].unique()
```

```python
# 빈도 수 확인 value_counts(): 각 항목별 갯수
df['Bank'].value_counts()
```

```python
# 상대 빈도 : normalize = True
df['Bank'].value_counts(normalize=True)
```
