## **Pandas Profiling**
- 데이터프레임을 빠르고 쉽게 요약 분석할 수 있도록 해주는 자동화된 탐색적 데이터 분석(EDA) 도구
- 데이터의 각 열에 대한 기술 통계와 시각화, 결측치 정보, 상관관계 등의 분석 결과를 자동으로 생성
- 데이터 분석 초기에 데이터를 파악하고 이상치를 발견하는 데 큰 도움

#### **YData Profiling의 주요 기능**
- **종합적인 데이터 요약:** 데이터의 변수별로 분포, 평균, 최솟값, 최댓값, 표준편차 등의 통계 요약 정보를 제공함.
- **결측치 분석:** 각 열의 결측값 수와 패턴을 시각적으로 보여주어, 데이터 정제에 필요한 정보를 쉽게 확인할 수 있음.
- **상관관계 분석:** 변수 간 상관관계를 매트릭스 형태로 제공하며, 변수 간 관계를 시각화함.
- **대용량 데이터 지원:** 작은 데이터셋뿐만 아니라 대용량 데이터셋에서도 사용할 수 있으며, Spark DataFrame도 지원함.
- **인터랙티브 보고서:** 분석 결과는 HTML 파일로 저장할 수 있으며, Jupyter 노트북에서 바로 시각화된 보고서를 출력할 수 있음.

### **판다스 프로파일링**

```python
# ydata 프로파일링 설치하고, 재시작
!pip install ydata-profiling
```

```python
# ydata 프로파일링 임포트 하기
import ydata_profiling as yp
# from ydata_profiling import ProfileReport
```

```python
# ProfileReport 객채 로드 하기 및 리포트 생성
profile = yp.ProfileReport(df, title="Profiling Report")
```

```python
# HTML 파일로 저장
profile.to_file("output.html")
```

```python
# Jupyter 노트북에서 리포트 출력하기
profile.to_notebook_iframe()
```
