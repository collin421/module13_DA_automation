## **데이터 합치기**

#### **1. 시간 구간이 다른 데이터 합치기**
- 주식 데이터를 월별 또는 연도별로 나누어 수집한 경우, 이를 결합해 전체 데이터를 분석할 수 있도록 통합해야 함.
- 예시: 1월과 2월의 주식 거래 데이터를 각각 다른 데이터프레임으로 가지고 있을 때, 이를 합쳐서 1~2월 동안의 트렌드를 분석할 필요가 있음.

#### **2. 다른 변수(열)를 가진 데이터 합치기**
- 한 데이터프레임에는 가격 데이터(종가, 시가)가 있고, 다른 데이터프레임에는 거래량 데이터가 있는 경우, 이를 결합해 가격과 거래량의 관계를 분석할 수 있음.
- 예시: 한 데이터프레임에 '종가', '시가'만 있고, 다른 데이터프레임에 '거래량'만 있을 때, 이를 합쳐 종가와 거래량 간의 상관관계를 분석함.

#### **데이터 결합의 이유**
- **1) 전체적인 패턴 분석:**
    - 예를 들어 1월과 2월 데이터를 결합함으로써 더 긴 기간 동안의 종가 변동과 거래량 패턴을 확인할 수 있음. 이는 특정 달이나 기간에만 집중하는 것보다 더 포괄적인 인사이트를 제공함.
- **2) 종합적 상관관계 분석:**
    - 시가, 종가, 거래량 같은 여러 변수들을 결합해 상관관계를 분석할 수 있음. 이를 통해 주가 변동에 따른 거래량 변화, 특정 종목의 거래 트렌드 등을 파악할 수 있음.
- **3) 데이터의 완성도 증가:**
    - 분리된 데이터를 결합함으로써 분석에 필요한 모든 변수를 하나의 데이터프레임에 통합할 수 있음. 이를 통해 다양한 분석을 한 번에 진행할 수 있고, 시각화도 용이해짐.

### **1) concat()** : 두 개 이상의 Pandas 데이터프레임이나 시리즈를 결합하는 데 사용됨. 
- 데이터는 행(axis=0) 또는 열(axis=1)을 기준으로 결합할 수 있음
- ignore_index: True로 설정하면 기존 인덱스를 무시하고 새로운 인덱스를 부여함

```python
import pandas as pd
```

#### **1.두개 데이터 합치기**

```python
# 데이터 불러오기
df1 = pd.read_csv('07_stock_concat1.csv', index_col=0)
df1
```

```python
# 두번째 데이터 불러오기
df2 = pd.read_csv('07_stock_concat2.csv', index_col=0)
```

```python
#데이터 합치기
df_concat = pd.concat([df1, df2], axis=0)
df_concat
```

#### **2. 여러개 데이터 합치기(for문)**

```python
# for문을 활용해서 4분기 데이터 모두 합치기

# 빈 데이터 프레임 생성
df_concat2 = pd.DataFrame()

# 1~4까지 반복
for i in range(1,5):
    # 데이터 불러오기
    dfn = pd.read_csv(f'07_stock_concat{i}.csv', index_col=0)
    
    # 기존 데이터 + 불러온 데이터 합쳐서 새로운 데이터 생성
    df_concat2 = pd.concat([df_concat2, dfn], axis=0, ignore_index=True)
    
    print(f'07_stock_concat{i} 파일 병합')

df_concat2
```

#### **3. 추가 컬럼 합치기**

```python
# 데이터 불러오기
extra_df = pd.read_csv('07_stock_extra.csv', index_col=0)
extra_df
```

```python
# 열 방향으로 합치기
# index가 같아야지 합쳐짐, reset_index(drop=True) 옵션
df_concat3 = pd.concat([df_concat2, extra_df], axis=1)
df_concat3
```

### **2) merge()** : 두 개 이상의 데이터프레임을 공통 열 또는 인덱스를 기준으로 병합할 때 사용됨
- **pd.merge(left, right, how='inner', on=None, left_index=False)**
    - left, right: 병합할 두 개의 데이터프레임.
    - how: 결합 방식 지정 ('left', 'right', 'outer', 'inner').
        - 'inner' (기본값): 공통된 데이터만 결합.
        - 'left': 왼쪽 데이터프레임의 모든 데이터를 유지.
        - 'right': 오른쪽 데이터프레임의 모든 데이터를 유지.
        - 'outer': 양쪽 데이터프레임의 모든 데이터를 결합.
    - on: 병합 기준이 되는 열. 공통된 열 이름이 있을 때 사용.
    - left_index, right_index: 인덱스를 기준으로 병합할 경우 True로 설정

#### **1.inner 방식 병합 (공통된 데이터만 결합)**

```python
# 데이터 불러오기
df_주가 = pd.read_csv('07_stock_merge_주가.csv', index_col=0)
df_재무정보 = pd.read_csv('07_stock_merge_재무정보.csv', index_col=0)
print(df_주가.shape, df_재무정보.shape)
```

```python
# '종목코드'를 기준으로 병합 (inner)
df_merged_inner = pd.merge(df_주가, df_재무정보, on='종목코드', how='inner')
df_merged_inner
```

#### **2. outer 방식 병합 (모든 데이터 결합)**
- 두 데이터프레임의 모든 행을 결합하며, 공통되지 않는 부분은 NaN으로 표시됨.

```python
# '종목코드'를 기준으로 병합 (inner)
df_merged_outer = pd.merge(df_주가, df_재무정보, on='종목코드', how='outer')
df_merged_outer
```

#### **merge 실습**
**1) 고객 데이터와 구매 데이터를 결합하여 고객별 총 구매금액 및 구매수량 데이터 만들기**</br>
**2) 연령열 추가하기**</br>

- 데이터 불러오기
    - 04_store_member.csv : 회원 정보
    - 04_store_product_1.csv : 회원들의 구매 정보
- 04_store_product_1.csv, 고객별 구매 금액, 구매 수량 합계 테이블 만들기
- 연령열 만들기
- 성별별 총 구매 금액, 구매 수량 통계 도출하기

```python
# 데이터 불러오기
df_member = pd.read_csv('./04_store_member.csv')
print(df_member.shape)
df_member.head(3)
```

```python
# 데이터 불러오기
df_product = pd.read_csv('./04_store_product_1.csv')
print(df_product.shape)
df_product.head(3)
```

#### **회원 번호를 기준으로 구매 금액, 수량 합계 테이블 만들기**

```python
# 회원 번호를 기준으로 구매 금액, 수량 합계 구하기
df_product2 = df_product.groupby('회원번호')[['구매금액', '구매수량']].sum()
df_product2 
```

#### **inner join**

```python
# 첫 번째 데이터의 회원 번호를 기준으로, 첫번째 데이터에 있는 데이터만 병합
df_inner = pd.merge(df_product2, df_member, on='회원번호', how='inner')
```

#### **right join**

```python
# 두 번째 데이터의 회원 번호를 기준으로 병합, 결측행이 많음
df_right = pd.merge(df_product2, df_member, on='회원번호', how='right')
```

#### **outer join**

```python
df_outer = pd.merge(df_product2, df_member, on='회원번호', how='outer')
```

```python
# 연령열 만들기
df_inner['연령'] = 2024 - df_inner['생년']
df_inner.head()
```

```python
# 성별에 따른 구매 금액 및 구매 수량
df_inner[['구매금액', '구매수량', '성별', '연령']].groupby('성별').sum()
```
