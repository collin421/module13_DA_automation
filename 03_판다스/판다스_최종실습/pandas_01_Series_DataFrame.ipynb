## **1) Series**
- index(순서)와 value(값)로 구성되어 있음
- 1차원 데이터를 다루는 구조임. 인덱스와 값으로 이루어져 있으며, 엑셀의 한 열과 유사함.
- 각 값에 고유한 인덱스가 있어, 데이터를 쉽게 조회하거나 연산할 수 있음.

### **Series 생성 방법**
- pd.Series()를 사용하여 생성 가능함.

#### **1. 리스트로 Series 생성**
- 리스트를 이용해 Series를 생성할 수 있음. 인덱스를 명시하지 않으면 기본적으로 0부터 시작하는 숫자가 인덱스로 자동 설정됨.

```python
# 리스트로 Series 생성
매출액 = pd.Series([1500, 2000, 2500, 3000, 3500, 4000])
매출액
```

#### **리스트 + 인덱스 지정**
- 리스트 외에도, 인덱스를 직접 지정하여 Series를 생성할 수 있음

```python
# 리스트와 인덱스 지정하여 Series 생성
매출액 = pd.Series([1500, 2000, 2500, 3000, 3500, 4000], index=['1월', '2월', '3월', '4월', '5월', '6월'])
매출액
```

#### **딕셔너리로 Series 생성**
- 딕셔너리 데이터 구조도 Series로 변환할 수 있음. 이 경우, 딕셔너리의 키가 인덱스가 되고, 값이 Series의 데이터가 됨.

```python
# 딕셔너리로 Series 생성
매출액 = pd.Series({'1월': 1500,
                 '2월': 2000,
                 '3월': 2500,
                 '4월': 3000,
                 '5월': 3500,
                 '6월': 4000,})
매출액
```

#### **넘파이 배열로 Series 생성**
- 넘파이 배열을 사용하여 Series를 만들 수 있으며, 넘파이 배열은 대용량 데이터 처리에 적합함.

```python
import numpy as np

# 넘파이 배열로 Series 생성
매출액 = pd.Series(np.array([1500, 2000, 2500, 3000, 3500, 4000]))
매출액
```

#### **Series 데이터 통계 지표**
- sum() : 합계
- mean() : 평균
- median() : 중앙값
- min() 최소값, max() 최댓값

```python
print(매출액.sum())
print(매출액.mean())
print(매출액.median())
print(매출액.min(), 매출액.max())
```

## **2) DataFrame**
- DataFrame은 판다스의 2차원 데이터 구조로, 행과 열로 이루어진 데이터 프레임임.
- 엑셀의 표와 유사한 구조로, 여러 개의 Series가 모여 행(row)과 열(column)을 이루 며 데이터를 구성함
- index(순서), column(속성), value(값)로 구성되어 있음

#### **1.딕셔너리로 DataFrame 생성**
- 딕셔너리를 사용하여 쉽게 DataFrame을 생성할 수 있음. 딕셔너리의 키는 열(column) 이름이 되고, 각 값은 해당 열의 데이터가 됨.

```python
# 딕셔너리로 DataFrame 생성
데이터 = {
    '상품명': ['A제품', 'B제품', 'C제품', 'D제품'],
    '매출액': [1500, 2000, 2500, 3000],
    '판매수량': [100, 150, 200, 250]
}

df = pd.DataFrame(데이터)
df
```

#### **리스트로 DataFrame 생성**
- 리스트의 리스트 구조를 사용하여 DataFrame을 생성할 수 있음. 각 내부 리스트는 하나의 행(row)을 나타냄.

```python
# 리스트로 DataFrame 생성
df2 = pd.DataFrame([
    ['A제품', 1500, 100],
    ['B제품', 2000, 150],
    ['C제품', 2500, 200],
    ['D제품', 3000, 250]
], columns=['상품명', '매출액', '판매수량'])

df2
```

### **DataFrame 다루기**

```python
# DataFrame 생성
data = {
    '종목명': ['삼성전자', 'SK하이닉스', 'NAVER', '카카오', 'LG화학',],
    '종목코드': ['005930', '000660', '035420', '035720', '051910'],
    '현재가': [70000, 120000, 380000, 135000, 800000],
    '거래량': [1000000, 800000, 500000, 600000, 300000],
    '지역': ['서울', '서울', '성남', '성남', '서울']
}

df = pd.DataFrame(data)
```

#### **set_index()** : 특정 열을 인덱스로 설정하는 데 사용됨. 
- 인덱스는 행의 식별자 역할을 하며, 데이터 조작 및 조회에 중요한 역할을 함.
- set_index('열이름'): 지정한 열을 인덱스로 설정.
- drop=False: 인덱스로 설정한 열을 데이터프레임에서 삭제하지 않고 유지.
- inplace=True: 원본 데이터프레임에 바로 적용

```python
# '종목코드'를 인덱스로 설정 (drop=False)
df_with_index = df.set_index('종목코드', drop=False)
df_with_index
```

```python
# '종목코드'를 인덱스로 설정 (drop=True)
df_with_index2 = df.set_index('종목코드', drop=True)
df_with_index2
```

```python
# 기존 데이터 프레임에 반영하기
df.set_index('종목코드', drop=True, inplace=True)
df
```

#### **reset_index()** 
- : 인덱스를 기본값(정수형 인덱스)으로 되돌리는 데 사용됨. 기존 인덱스는 열로 변환되고, 정수형 인덱스가 새로 부여됨.
- drop=True: 기존 인덱스를 열로 변환하지 않고 제거.
- inplace=True: 원본 데이터프레임에 바로 적용.

```python
# index열 열로 전환
df.reset_index(inplace=True)

# index열 삭제
# df.reset_index(inplace=True, drop=True)

df
```

#### **rename(): 행 또는 열 이름 변경**
- 데이터프레임의 행 인덱스 또는 열 이름을 변경할 때 사용됨. **딕셔너리**를 사용하여 원하는 이름으로 변경할 수 있음
- columns={'old_name': 'new_name'}: 열 이름을 변경.
- index={'old_name': 'new_name'}: 행 이름(인덱스)을 변경.
- inplace=True: 원본 데이터프레임에 바로 적용

```python
df.set_index('종목코드', drop=True, inplace=True)
df
```

```python
# 열 이름 변경
df.rename(columns={'종목명': '회사명', '현재가': '가격'}, inplace=True)
df
```

```python
# 행 인덱스(종목코드) 이름 변경
df.rename(index={'005930': '삼성전자', '000660': 'SK하이닉스'}, inplace=True)
df
```

#### **drop(): 행 또는 열 삭제**
- axis=0: 행을 삭제.
- axis=1: 열을 삭제.
- inplace=True: 원본 데이터프레임에 바로 적용

```python
# 회사명 열 삭제
df2 = df.drop('회사명', axis=1)
df2
```

```python
# 종목코드 SK하이닉스, 삼성전자 삭제
df3 = df.drop(labels=['SK하이닉스','삼성전자'] , axis=0)
df3
```

### **3. iloc[]와 loc[]**
- 1. iloc[]: 숫자 인덱스(정수 위치) 사용하여 데이터프레임의 특정 위치에 있는 행과 열을 선택하는 방식임
- 2. loc[]는 인덱스 이름 또는 열 이름을 사용하여 데이터를 선택하는 방식임. 즉, 행과 열의 이름(레이블) 사용

#### **사용법 iloc[행위치 , 열위치]** 
-  iloc[row_index, column_index]
-  start : end : step

```python
# iloc[] : 첫번째 행 출력하기
df.iloc[0] # 첫번째 행 출력
```

```python
# 세번째 행 출력
df.iloc[2]
```

```python
# 첫번째 부터 5번째 행까지 출력
df.iloc[0:5] # 0은 생략 가능
# df.iloc[:5]
```

```python
# 10번째 행까지 출력, 2step씩 출력
df.iloc[:10:2]
```

```python
# 5행 부터 끝까지 출력
df.iloc[5:]
```

```python
# 3행부터 마지막에서 두번째 행까지 출력
df.iloc[2:-2]
```

### 사용법 **loc[]** 
- 인덱스 이름 또는 열 이름을 사용하여 데이터를 선택하는 방식임. 
- 행과 열을 선택할 때 정수 인덱스가 아니라 **이름(레이블)**을 사용
- **loc[행 인덱스 이름, 열 인덱스 이름]**
- **start : end : step**

```python
# 종목명 열로 인덱스 재 설정, 기존 열 삭제
df2 = df.set_index('종목명', drop=True)
df2
```

```python
# 특정 행 선택, 행 인덱스 이름으로 인덱싱
df2.loc['삼성전자']
```

```python
# 여러행 선택, 떨어져 있는 행
df2.loc[['삼성전자', 'NAVER']]
```

```python
# 여러행 선택 : 연속되는 여러행 선택
df2.loc['삼성전자':'현대차']
```

```python
# 모든행의 특정 열 선택, 인덱스도 함게 출력
df2.loc[:, '지역']
```

```python
# 모든행의 복수의 열 선택 - 떨어져 있는
df2.loc[:, ['현재가','지역']]
```

```python
# 모든행의 복수의 열 선택 - 연속되는
df2.loc[:, '현재가':'지역']
```

```python
# 행 열 동시 선택
# df2.loc['삼성전자', '지역'] # 인덱스 삼성전자, 지역열 
# df2.loc[['삼성전자', 'NAVER'], '지역'] # 인덱스 삼성전자, 네이버의 지역
df2.loc['삼성전자':'현대차', '현재가':'지역'] # 삼성전자~현대차까지, 현재~지역열 출력
```
